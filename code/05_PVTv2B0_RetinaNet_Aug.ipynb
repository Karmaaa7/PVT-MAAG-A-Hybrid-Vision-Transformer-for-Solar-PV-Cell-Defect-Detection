# CSE465 — PVTv2-B0 + RetinaNet (Augmented)

Final augmentation experiment for PVTv2-B0 + FPN + RetinaNet.

This version is designed specifically to maximize the chance of beating the
successful NoAug baseline while avoiding the failure mode observed with strong
geometric augmentation.

## Final strategy
Use **photometric-dominant augmentation** plus geometry-preserving flips:

- brightness ±15%, p=0.50
- contrast ±10%, p=0.50
- mild Gaussian blur, p=0.10
- mild Gaussian noise, p=0.15
- horizontal flip, p=0.50
- vertical flip, p=0.50

Disabled:
- rotation
- translation
- scaling
- shear
- Mosaic
- MixUp
- CutMix
- Copy-Paste
- erasing

This preserves defect geometry, especially the elongated
horizontal/vertical dislocation classes.

Validation and test remain unaugmented.

## 8. Dataset class — photometric-dominant augmentation

Training images are letterboxed to 640×640, then receive only:
- brightness variation
- contrast variation
- mild blur
- mild Gaussian noise
- horizontal/vertical flips

Only flips modify box coordinates, and torchvision v2 transforms the boxes
together with the image.

Because there is no rotation, translation, crop, scale or shear, box counts
must remain exactly unchanged.

Validation/test use deterministic letterboxing only.

````python
IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png",
    ".bmp", ".tif", ".tiff",
}


def letterbox_image(
    image,
    boxes,
    size=640,
    fill=114,
):
    h, w = image.shape[:2]

    scale = min(
        size / w,
        size / h,
    )

    new_w = int(
        round(w * scale)
    )

    new_h = int(
        round(h * scale)
    )

    resized = cv2.resize(
        image,
        (new_w, new_h),
        interpolation=cv2.INTER_LINEAR,
    )

    pad_x = (
        size - new_w
    ) // 2

    pad_y = (
        size - new_h
    ) // 2

    canvas = np.full(
        (size, size, 3),
        fill,
        dtype=np.uint8,
    )

    canvas[
        pad_y:pad_y + new_h,
        pad_x:pad_x + new_w,
    ] = resized

    boxes = np.asarray(
        boxes,
        dtype=np.float32,
    ).reshape(-1, 4)

    if len(boxes):
        boxes = boxes.copy()

        boxes[:, [0, 2]] = (
            boxes[:, [0, 2]]
            * scale
            + pad_x
        )

        boxes[:, [1, 3]] = (
            boxes[:, [1, 3]]
            * scale
            + pad_y
        )

    return canvas, boxes


train_transform = v2.Compose([
    v2.RandomApply(
        [
            v2.ColorJitter(
                brightness=AUGMENTATION["brightness"],
                contrast=0.0,
                saturation=0.0,
                hue=0.0,
            )
        ],
        p=AUGMENTATION["brightness_p"],
    ),

    v2.RandomApply(
        [
            v2.ColorJitter(
                brightness=0.0,
                contrast=AUGMENTATION["contrast"],
                saturation=0.0,
                hue=0.0,
            )
        ],
        p=AUGMENTATION["contrast_p"],
    ),

    v2.RandomApply(
        [
            v2.GaussianBlur(
                kernel_size=AUGMENTATION[
                    "gaussian_blur_kernel"
                ],
                sigma=(
                    AUGMENTATION[
                        "gaussian_blur_sigma_min"
                    ],
                    AUGMENTATION[
                        "gaussian_blur_sigma_max"
                    ],
                ),
            )
        ],
        p=AUGMENTATION["gaussian_blur_p"],
    ),

    v2.RandomHorizontalFlip(
        p=AUGMENTATION["fliplr"]
    ),

    v2.RandomVerticalFlip(
        p=AUGMENTATION["flipud"]
    ),
])


class PVDetectionDataset(Dataset):
    def __init__(
        self,
        dataset_dir,
        split,
        image_size=640,
        augment=False,
    ):
        self.image_dir = (
            Path(dataset_dir)
            / "images"
            / split
        )

        self.label_dir = (
            Path(dataset_dir)
            / "labels"
            / split
        )

        self.split = split
        self.image_size = image_size
        self.augment = augment

        self.images = sorted(
            p
            for p in self.image_dir.iterdir()
            if p.suffix.lower()
            in IMAGE_EXTENSIONS
        )

    def __len__(self):
        return len(self.images)

    def _load_raw(self, idx):
        image_path = self.images[idx]

        label_path = (
            self.label_dir
            / f"{image_path.stem}.txt"
        )

        image = cv2.imread(
            str(image_path)
        )

        if image is None:
            raise RuntimeError(
                f"Could not read: {image_path}"
            )

        image = cv2.cvtColor(
            image,
            cv2.COLOR_BGR2RGB,
        )

        h, w = image.shape[:2]

        boxes = []
        labels = []

        if label_path.exists():
            text = (
                label_path
                .read_text()
                .strip()
            )

            if text:
                for line in text.splitlines():
                    (
                        cls_id,
                        xc,
                        yc,
                        bw,
                        bh,
                    ) = map(
                        float,
                        line.split(),
                    )

                    x1 = (
                        xc - bw / 2
                    ) * w

                    y1 = (
                        yc - bh / 2
                    ) * h

                    x2 = (
                        xc + bw / 2
                    ) * w

                    y2 = (
                        yc + bh / 2
                    ) * h

                    boxes.append(
                        [x1, y1, x2, y2]
                    )

                    labels.append(
                        int(cls_id)
                    )

        boxes = np.asarray(
            boxes,
            dtype=np.float32,
        ).reshape(-1, 4)

        labels = np.asarray(
            labels,
            dtype=np.int64,
        ).reshape(-1)

        return (
            image,
            boxes,
            labels,
            str(image_path),
        )

    def __getitem__(self, idx):
        (
            image,
            boxes,
            labels,
            image_path,
        ) = self._load_raw(idx)

        image, boxes = letterbox_image(
            image,
            boxes,
            size=self.image_size,
        )

        image = tv_tensors.Image(
            torch.from_numpy(
                np.ascontiguousarray(
                    image
                )
            ).permute(2, 0, 1)
        )

        boxes = tv_tensors.BoundingBoxes(
            torch.as_tensor(
                boxes,
                dtype=torch.float32,
            ).reshape(-1, 4),
            format="XYXY",
            canvas_size=(
                self.image_size,
                self.image_size,
            ),
        )

        target = {
            "boxes": boxes,
            "labels": torch.as_tensor(
                labels,
                dtype=torch.int64,
            ),
        }

        original_box_count = len(
            target["boxes"]
        )

        if self.augment:
            image, target = train_transform(
                image,
                target,
            )

        image = (
            image
            .as_subclass(torch.Tensor)
            .to(torch.float32)
            / 255.0
        )

        if (
            self.augment
            and random.random()
            < AUGMENTATION["gaussian_noise_p"]
        ):
            image = torch.clamp(
                image
                + torch.randn_like(image)
                * AUGMENTATION[
                    "gaussian_noise_std"
                ],
                0.0,
                1.0,
            )

        boxes_out = (
            target["boxes"]
            .as_subclass(torch.Tensor)
            .to(torch.float32)
        )

        labels_out = (
            target["labels"]
            .to(torch.int64)
        )

        if len(boxes_out) != original_box_count:
            raise RuntimeError(
                f"Unexpected box-count change: "
                f"{image_path}"
            )

        if len(boxes_out):
            if not torch.isfinite(
                boxes_out
            ).all():
                raise RuntimeError(
                    f"Non-finite box: "
                    f"{image_path}"
                )

            if not (
                (
                    boxes_out[:, 2]
                    > boxes_out[:, 0]
                ).all()
                and (
                    boxes_out[:, 3]
                    > boxes_out[:, 1]
                ).all()
            ):
                raise RuntimeError(
                    f"Invalid box: "
                    f"{image_path}"
                )

        target_out = {
            "boxes": boxes_out,
            "labels": labels_out,
            "image_id": torch.tensor(
                [idx],
                dtype=torch.int64,
            ),
        }

        return (
            image,
            target_out,
            image_path,
        )


def collate_fn(batch):
    images, targets, paths = zip(*batch)

    return (
        list(images),
        list(targets),
        list(paths),
    )


def seed_worker(worker_id):
    worker_seed = (
        torch.initial_seed()
        % 2**32
    )

    np.random.seed(worker_seed)
    random.seed(worker_seed)


loader_generator = torch.Generator()
loader_generator.manual_seed(SEED)


train_dataset = PVDetectionDataset(
    DATASET_DIR,
    "train",
    IMAGE_SIZE,
    augment=True,
)

val_dataset = PVDetectionDataset(
    DATASET_DIR,
    "val",
    IMAGE_SIZE,
    augment=False,
)

test_dataset = PVDetectionDataset(
    DATASET_DIR,
    "test",
    IMAGE_SIZE,
    augment=False,
)


train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    collate_fn=collate_fn,
    worker_init_fn=seed_worker,
    generator=loader_generator,
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    collate_fn=collate_fn,
    worker_init_fn=seed_worker,
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    collate_fn=collate_fn,
    worker_init_fn=seed_worker,
)

print(
    "Train:", len(train_dataset),
    "(augmentation ON)"
)

print(
    "Val  :", len(val_dataset),
    "(augmentation OFF)"
)

print(
    "Test :", len(test_dataset),
    "(augmentation OFF)"
)

print(
    "Effective batch:",
    EFFECTIVE_BATCH
)
````

### 8.1 Mandatory augmentation preflight

This configuration does not crop or geometrically deform boxes, so the
preflight requires **100% box-count retention**.

It checks 1,000 random training samples before the long run.

````python
from matplotlib.patches import Rectangle

def augmentation_preflight(
    dataset,
    sample_count=1000,
):
    selector = random.Random(SEED)

    indices = [
        selector.randrange(
            len(dataset)
        )
        for _ in range(sample_count)
    ]

    raw_counts = np.zeros(
        NUM_CLASSES,
        dtype=np.int64,
    )

    aug_counts = np.zeros(
        NUM_CLASSES,
        dtype=np.int64,
    )

    invalid = 0

    for idx in indices:
        (
            _,
            raw_boxes,
            raw_labels,
            _,
        ) = dataset._load_raw(idx)

        raw_counts += np.bincount(
            raw_labels,
            minlength=NUM_CLASSES,
        )

        image, target, _ = dataset[idx]

        boxes = (
            target["boxes"]
            .cpu()
            .numpy()
        )

        labels = (
            target["labels"]
            .cpu()
            .numpy()
        )

        aug_counts += np.bincount(
            labels,
            minlength=NUM_CLASSES,
        )

        if len(boxes) != len(raw_boxes):
            invalid += 1
            continue

        if len(boxes):
            if not np.isfinite(
                boxes
            ).all():
                invalid += 1
                continue

            if not (
                (
                    boxes[:, 2]
                    > boxes[:, 0]
                ).all()
                and (
                    boxes[:, 3]
                    > boxes[:, 1]
                ).all()
            ):
                invalid += 1

    print("=" * 70)
    print("AUGMENTATION PREFLIGHT")
    print("=" * 70)
    print(
        "Samples checked:",
        sample_count,
    )
    print(
        "Invalid samples:",
        invalid,
    )

    if invalid != 0:
        raise RuntimeError(
            "Augmentation preflight failed. "
            "Do NOT start training."
        )

    print(
        "\nClass-wise annotation retention:"
    )

    for cls_id, name in enumerate(
        CLASS_NAMES
    ):
        raw_n = int(
            raw_counts[cls_id]
        )

        aug_n = int(
            aug_counts[cls_id]
        )

        if raw_n == 0:
            print(
                f"{name:25s}: "
                "not sampled"
            )
            continue

        print(
            f"{name:25s}: "
            f"{aug_n}/{raw_n} "
            f"({aug_n/raw_n:.1%})"
        )

        if aug_n != raw_n:
            raise RuntimeError(
                f"Annotation count changed "
                f"for {name}. "
                "Do NOT train."
            )

    print(
        "\nPreflight PASSED."
    )


augmentation_preflight(
    train_dataset,
    sample_count=1000,
)


fig, axes = plt.subplots(
    2,
    4,
    figsize=(20, 10),
)

axes = axes.ravel()

preview_rng = random.Random(
    SEED + 1
)

for ax in axes:
    idx = preview_rng.randrange(
        len(train_dataset)
    )

    image, target, _ = (
        train_dataset[idx]
    )

    ax.imshow(
        image
        .permute(1, 2, 0)
        .cpu()
        .numpy()
        .clip(0, 1)
    )

    for box, cls_id in zip(
        target["boxes"].numpy(),
        target["labels"].numpy(),
    ):
        x1, y1, x2, y2 = box

        ax.add_patch(
            Rectangle(
                (x1, y1),
                x2 - x1,
                y2 - y1,
                fill=False,
                linewidth=1.5,
            )
        )

        ax.text(
            x1,
            max(10, y1),
            CLASS_NAMES[
                int(cls_id)
            ],
            fontsize=7,
            bbox=dict(
                facecolor="white",
                alpha=0.65,
                edgecolor="none",
            ),
        )

    ax.axis("off")

plt.suptitle(
    "PVTv2-B0 final augmentation preview",
    fontsize=14,
)

plt.tight_layout()
plt.show()

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

print(
    "RNGs reset after preflight."
)
````

## 9. Build PVTv2-B0 + FPN + RetinaNet

`timm` supplies the pretrained PVTv2-B0 backbone in feature-extraction mode.

The last three hierarchical feature stages are sent to an FPN, followed by a RetinaNet detection head.

The code reads the feature-channel dimensions directly from `timm` instead of hard-coding them.

````python
class PVTv2B0FPN(nn.Module):
    def __init__(self, pretrained=True, out_channels=256):
        super().__init__()

        self.body = timm.create_model(
            "pvt_v2_b0",
            pretrained=pretrained,
            features_only=True,
            out_indices=(1, 2, 3),
        )

        feature_channels = self.body.feature_info.channels()
        feature_reductions = self.body.feature_info.reduction()

        print("PVTv2 feature channels:", feature_channels)
        print("PVTv2 feature strides :", feature_reductions)

        self.fpn = FeaturePyramidNetwork(
            in_channels_list=feature_channels,
            out_channels=out_channels,
            extra_blocks=LastLevelP6P7(
                feature_channels[-1],
                out_channels,
            ),
        )

        self.out_channels = out_channels

    def forward(self, x):
        features = self.body(x)

        features = OrderedDict(
            (str(i), feat)
            for i, feat in enumerate(features)
        )

        return self.fpn(features)


def build_model():
    backbone = PVTv2B0FPN(
        pretrained=True,
        out_channels=256,
    )

    model = RetinaNet(
        backbone=backbone,
        num_classes=NUM_CLASSES,
        min_size=IMAGE_SIZE,
        max_size=IMAGE_SIZE,
        image_mean=[0.485, 0.456, 0.406],
        image_std=[0.229, 0.224, 0.225],
        score_thresh=0.001,
        detections_per_img=100,
    )

    return model


model = build_model().to(DEVICE)

total_params = sum(
    p.numel()
    for p in model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print(
    f"Total parameters    : "
    f"{total_params:,} ({total_params/1e6:.3f} M)"
)

print(
    f"Trainable parameters: "
    f"{trainable_params:,} ({trainable_params/1e6:.3f} M)"
)
````

## 11. Validation helpers

````python
def move_targets_to_device(targets):
    return [
        {
            k: v.to(
                DEVICE,
                non_blocking=True,
            )
            for k, v in t.items()
        }
        for t in targets
    ]


@torch.no_grad()
def compute_validation_loss(model, loader):
    # Torchvision detection models return losses in train mode.
    # PVTv2-B0 in timm has no random augmentation here;
    # gradients remain disabled by @torch.no_grad().
    model.train()

    total_cls = 0.0
    total_box = 0.0
    total_images = 0

    start = time.perf_counter()

    for images, targets, _ in loader:
        images = [
            img.to(
                DEVICE,
                non_blocking=True,
            )
            for img in images
        ]

        targets = move_targets_to_device(targets)

        with autocast_context():
            losses = model(
                images,
                targets,
            )

        batch_n = len(images)

        total_cls += (
            float(
                losses["classification"].detach()
            )
            * batch_n
        )

        total_box += (
            float(
                losses["bbox_regression"].detach()
            )
            * batch_n
        )

        total_images += batch_n

    torch.cuda.synchronize()

    seconds = (
        time.perf_counter()
        - start
    )

    model.eval()

    cls_loss = (
        total_cls
        / max(total_images, 1)
    )

    box_loss = (
        total_box
        / max(total_images, 1)
    )

    return {
        "classification_loss": cls_loss,
        "bbox_regression_loss": box_loss,
        "total_loss": cls_loss + box_loss,
        "seconds": seconds,
    }


@torch.no_grad()
def evaluate_map(model, loader):
    model.eval()

    metric = MeanAveragePrecision(
        box_format="xyxy",
        iou_type="bbox",
        class_metrics=False,
    )

    start = time.perf_counter()

    for images, targets, _ in loader:
        images_gpu = [
            img.to(
                DEVICE,
                non_blocking=True,
            )
            for img in images
        ]

        with autocast_context():
            outputs = model(images_gpu)

        preds = [
            {
                "boxes": out["boxes"].detach().cpu(),
                "scores": out["scores"].detach().cpu(),
                "labels": out["labels"].detach().cpu(),
            }
            for out in outputs
        ]

        gts = [
            {
                "boxes": t["boxes"].detach().cpu(),
                "labels": t["labels"].detach().cpu(),
            }
            for t in targets
        ]

        metric.update(preds, gts)

    torch.cuda.synchronize()

    seconds = (
        time.perf_counter()
        - start
    )

    result = metric.compute()

    return {
        "map50": float(result["map_50"]),
        "map75": float(result["map_75"]),
        "map50_95": float(result["map"]),
        "seconds": seconds,
    }
````

## 13. Train — live epoch/batch progress

The model is selected using validation **mAP@50-95**.

Early stopping:
- maximum = 50 epochs
- patience = 5

The live progress bar shows current batch loss and VRAM.

````python
from tqdm.auto import tqdm

for epoch in range(
    start_epoch,
    MAX_EPOCHS + 1,
):

    model.train()

    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()

    epoch_start = time.perf_counter()

    train_cls_sum = 0.0
    train_box_sum = 0.0
    train_images = 0

    optimizer.zero_grad(
        set_to_none=True
    )

    n_steps = len(train_loader)

    train_bar = tqdm(
        enumerate(train_loader),
        total=n_steps,
        desc=f"Epoch {epoch}/{MAX_EPOCHS}",
        dynamic_ncols=True,
    )

    for step, (
        images,
        targets,
        _,
    ) in train_bar:

        images = [
            img.to(
                DEVICE,
                non_blocking=True,
            )
            for img in images
        ]

        targets = move_targets_to_device(
            targets
        )

        group_start = (
            step // ACCUM_STEPS
        ) * ACCUM_STEPS

        group_end = min(
            group_start + ACCUM_STEPS,
            n_steps,
        )

        group_size = (
            group_end - group_start
        )

        with autocast_context():
            losses = model(
                images,
                targets,
            )

            cls_loss = losses[
                "classification"
            ]

            box_loss = losses[
                "bbox_regression"
            ]

            total_loss = (
                cls_loss + box_loss
            )

            if not torch.isfinite(
                total_loss
            ):
                raise RuntimeError(
                    f"Non-finite training loss at "
                    f"epoch {epoch}, step {step}. "
                    "Training stopped immediately."
                )

            scaled_loss = (
                total_loss
                / group_size
            )

        scaler.scale(
            scaled_loss
        ).backward()

        if step == group_end - 1:
            scaler.unscale_(optimizer)

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                GRAD_CLIP,
            )

            scaler.step(optimizer)
            scaler.update()

            optimizer.zero_grad(
                set_to_none=True
            )

        batch_n = len(images)

        train_cls_sum += (
            float(cls_loss.detach())
            * batch_n
        )

        train_box_sum += (
            float(box_loss.detach())
            * batch_n
        )

        train_images += batch_n

        train_bar.set_postfix({
            "loss": (
                f"{float(total_loss.detach()):.4f}"
            ),
            "cls": (
                f"{float(cls_loss.detach()):.4f}"
            ),
            "box": (
                f"{float(box_loss.detach()):.4f}"
            ),
            "VRAM": (
                f"{torch.cuda.memory_allocated()/1024**3:.1f}G"
            ),
        })

    torch.cuda.synchronize()

    train_seconds = (
        time.perf_counter()
        - epoch_start
    )

    train_cls = (
        train_cls_sum
        / train_images
    )

    train_box = (
        train_box_sum
        / train_images
    )

    train_total = (
        train_cls + train_box
    )

    print(
        "\nCalculating validation loss..."
    )

    val_loss_result = (
        compute_validation_loss(
            model,
            val_loader,
        )
    )

    print(
        "Calculating validation mAP..."
    )

    val_map_result = evaluate_map(
        model,
        val_loader,
    )

    scheduler.step()

    total_seconds = (
        train_seconds
        + val_loss_result["seconds"]
        + val_map_result["seconds"]
    )

    peak_vram_gb = (
        torch.cuda.max_memory_allocated()
        / 1024**3
    )

    record = {
        "epoch": epoch,

        "train/classification_loss":
            train_cls,

        "train/bbox_regression_loss":
            train_box,

        "train/total_loss":
            train_total,

        "val/classification_loss":
            val_loss_result[
                "classification_loss"
            ],

        "val/bbox_regression_loss":
            val_loss_result[
                "bbox_regression_loss"
            ],

        "val/total_loss":
            val_loss_result[
                "total_loss"
            ],

        "val/mAP50":
            val_map_result["map50"],

        "val/mAP75":
            val_map_result["map75"],

        "val/mAP50-95":
            val_map_result["map50_95"],

        "train_seconds":
            train_seconds,

        "val_loss_seconds":
            val_loss_result["seconds"],

        "val_map_seconds":
            val_map_result["seconds"],

        "total_seconds":
            total_seconds,

        "lr":
            optimizer.param_groups[0]["lr"],

        "peak_vram_GB":
            peak_vram_gb,
    }

    history.append(record)

    history_df = pd.DataFrame(
        history
    )

    history_df.to_csv(
        METRICS_DIR
        / "training_history.csv",
        index=False,
    )

    current_map = (
        val_map_result["map50_95"]
    )

    if (
        not np.isfinite(current_map)
        or current_map < 0
    ):
        raise RuntimeError(
            f"Invalid validation mAP50-95 "
            f"at epoch {epoch}: {current_map}. "
            "Checkpoint remains available from the "
            "previous completed epoch."
        )

    improved = (
        current_map > best_map
    )

    if improved:
        best_map = current_map
        best_epoch = epoch

        epochs_without_improvement = 0

        best_state = {
            "epoch": epoch,
            "best_map50_95": best_map,

            "model_state": {
                k: v.detach().cpu()
                for k, v
                in model.state_dict().items()
            },

            "class_names": CLASS_NAMES,
            "training_config":
                training_config,
        }

        torch.save(
            best_state,
            BEST_PT,
        )

    else:
        epochs_without_improvement += 1

    last_state = {
        "epoch": epoch,

        "model_state":
            model.state_dict(),

        "optimizer_state":
            optimizer.state_dict(),

        "scheduler_state":
            scheduler.state_dict(),

        "scaler_state":
            scaler.state_dict(),

        "best_map":
            best_map,

        "best_epoch":
            best_epoch,

        "epochs_without_improvement":
            epochs_without_improvement,

        "history":
            history,

        "training_config":
            training_config,
    }

    torch.save(
        last_state,
        LAST_PT,
    )

    print(
        "\n"
        + "=" * 70
    )

    print(
        f"EPOCH {epoch}/{MAX_EPOCHS} COMPLETE"
    )

    print("-" * 70)

    print(
        f"Train Loss    : "
        f"{train_total:.4f} "
        f"(Cls {train_cls:.4f} | "
        f"Box {train_box:.4f})"
    )

    print(
        f"Val Loss      : "
        f"{val_loss_result['total_loss']:.4f}"
    )

    print(
        f"Val mAP@50    : "
        f"{val_map_result['map50']:.4f}"
    )

    print(
        f"Val mAP@75    : "
        f"{val_map_result['map75']:.4f}"
    )

    print(
        f"Val mAP@50-95 : "
        f"{current_map:.4f}"
    )

    print("-" * 70)

    print(
        f"Train time    : "
        f"{train_seconds/60:.2f} min"
    )

    print(
        f"Val-loss time : "
        f"{val_loss_result['seconds']/60:.2f} min"
    )

    print(
        f"Val-mAP time  : "
        f"{val_map_result['seconds']/60:.2f} min"
    )

    print(
        f"Total epoch   : "
        f"{total_seconds/60:.2f} min"
    )

    print(
        f"Peak VRAM     : "
        f"{peak_vram_gb:.2f} GB"
    )

    print("-" * 70)

    print(
        f"Best epoch    : "
        f"{best_epoch}"
    )

    print(
        f"Best mAP50-95 : "
        f"{best_map:.4f}"
    )

    print(
        f"Early stopping: "
        f"{epochs_without_improvement}/{PATIENCE}"
    )

    print("=" * 70)

    if (
        epochs_without_improvement
        >= PATIENCE
    ):
        print(
            f"\nEarly stopping triggered "
            f"at epoch {epoch}."
        )

        break

print("\nTRAINING FINISHED")
print("Best epoch:", best_epoch)

print(
    "Best validation mAP@50-95:",
    round(best_map, 4),
)
````

## 18. Collect test predictions once

````python
@torch.no_grad()
def collect_predictions(
    model,
    loader,
):
    model.eval()

    predictions = []
    targets_all = []
    paths_all = []

    total_forward_seconds = 0.0
    total_images = 0

    for images, targets, paths in loader:
        images_gpu = [
            img.to(
                DEVICE,
                non_blocking=True,
            )
            for img in images
        ]

        torch.cuda.synchronize()
        start = time.perf_counter()

        with autocast_context():
            outputs = model(
                images_gpu
            )

        torch.cuda.synchronize()

        total_forward_seconds += (
            time.perf_counter()
            - start
        )

        total_images += len(images)

        for out in outputs:
            predictions.append({
                "boxes":
                    out["boxes"].detach().cpu(),

                "scores":
                    out["scores"].detach().cpu(),

                "labels":
                    out["labels"].detach().cpu(),
            })

        for target in targets:
            targets_all.append({
                "boxes":
                    target["boxes"].detach().cpu(),

                "labels":
                    target["labels"].detach().cpu(),
            })

        paths_all.extend(paths)

    inference_ms = (
        total_forward_seconds
        / max(total_images, 1)
        * 1000
    )

    return (
        predictions,
        targets_all,
        paths_all,
        inference_ms,
    )


(
    test_predictions,
    test_targets,
    test_paths,
    inference_ms,
) = collect_predictions(
    model,
    test_loader,
)

fps = (
    1000 / inference_ms
    if inference_ms > 0
    else np.nan
)

print(
    "Test images:",
    len(test_predictions),
)

print(
    f"Model forward time: "
    f"{inference_ms:.3f} ms/image"
)

print(
    f"FPS: {fps:.2f}"
)
````

## 20. Precision, Recall, F1, matched box IoU and Dice

Threshold metrics:
- confidence ≥ 0.25
- class-aware one-to-one matching
- TP IoU threshold = 0.50

IoU and Dice are **bounding-box overlap metrics**, not segmentation-mask metrics.

````python
def class_aware_matches(
    pred,
    gt,
    conf_threshold=0.25,
    iou_threshold=0.50,
):
    keep = (
        pred["scores"]
        >= conf_threshold
    )

    p_boxes = pred["boxes"][keep]
    p_labels = pred["labels"][keep]

    g_boxes = gt["boxes"]
    g_labels = gt["labels"]

    if (
        len(p_boxes) == 0
        or len(g_boxes) == 0
    ):
        return (
            [],
            p_boxes,
            p_labels,
        )

    ious = box_iou(
        p_boxes,
        g_boxes,
    )

    candidates = []

    for pi in range(
        len(p_boxes)
    ):
        for gi in range(
            len(g_boxes)
        ):
            if (
                int(p_labels[pi])
                != int(g_labels[gi])
            ):
                continue

            iou = float(
                ious[pi, gi]
            )

            if (
                iou
                >= iou_threshold
            ):
                candidates.append(
                    (
                        iou,
                        pi,
                        gi,
                    )
                )

    candidates.sort(
        reverse=True
    )

    used_pred = set()
    used_gt = set()
    matches = []

    for iou, pi, gi in candidates:
        if (
            pi in used_pred
            or gi in used_gt
        ):
            continue

        used_pred.add(pi)
        used_gt.add(gi)

        matches.append(
            (
                pi,
                gi,
                iou,
            )
        )

    return (
        matches,
        p_boxes,
        p_labels,
    )


overall_tp = 0
overall_fp = 0
overall_fn = 0

matched_ious = []

per_class_counts = defaultdict(
    lambda: {
        "tp": 0,
        "fp": 0,
        "fn": 0,
        "ious": [],
    }
)

for pred, gt in zip(
    test_predictions,
    test_targets,
):
    (
        matches,
        p_boxes,
        p_labels,
    ) = class_aware_matches(
        pred,
        gt,
        CONF_THRESHOLD,
        MATCH_IOU_THRESHOLD,
    )

    matched_p = {
        m[0]
        for m in matches
    }

    matched_g = {
        m[1]
        for m in matches
    }

    tp = len(matches)
    fp = len(p_boxes) - tp
    fn = len(gt["boxes"]) - tp

    overall_tp += tp
    overall_fp += fp
    overall_fn += fn

    for (
        pi,
        gi,
        iou,
    ) in matches:
        cls_id = int(
            gt["labels"][gi]
        )

        matched_ious.append(
            iou
        )

        per_class_counts[
            cls_id
        ]["tp"] += 1

        per_class_counts[
            cls_id
        ]["ious"].append(
            iou
        )

    for pi, cls_id in enumerate(
        p_labels.tolist()
    ):
        if pi not in matched_p:
            per_class_counts[
                int(cls_id)
            ]["fp"] += 1

    for gi, cls_id in enumerate(
        gt["labels"].tolist()
    ):
        if gi not in matched_g:
            per_class_counts[
                int(cls_id)
            ]["fn"] += 1


test_precision = (
    overall_tp
    / (overall_tp + overall_fp)
    if (
        overall_tp + overall_fp
    )
    else 0.0
)

test_recall = (
    overall_tp
    / (overall_tp + overall_fn)
    if (
        overall_tp + overall_fn
    )
    else 0.0
)

test_f1 = (
    2
    * test_precision
    * test_recall
    / (
        test_precision
        + test_recall
    )
    if (
        test_precision
        + test_recall
    )
    else 0.0
)

mean_iou = (
    float(
        np.mean(matched_ious)
    )
    if matched_ious
    else np.nan
)

matched_dices = [
    2 * iou / (1 + iou)
    for iou in matched_ious
]

mean_dice = (
    float(
        np.mean(matched_dices)
    )
    if matched_dices
    else np.nan
)

print(
    f"Precision : "
    f"{test_precision:.4f}"
)

print(
    f"Recall    : "
    f"{test_recall:.4f}"
)

print(
    f"F1        : "
    f"{test_f1:.4f}"
)

print(
    f"TP / FP / FN: "
    f"{overall_tp} / "
    f"{overall_fp} / "
    f"{overall_fn}"
)

print(
    f"Mean matched box IoU : "
    f"{mean_iou:.4f}"
)

print(
    f"Mean matched box Dice: "
    f"{mean_dice:.4f}"
)
````

## 21. Per-class test performance

````python
map95_by_class = {}
map50_by_class = {}

classes95 = (
    map_result["classes"]
    .cpu()
    .numpy()
    .astype(int)
)

values95 = (
    map_result["map_per_class"]
    .cpu()
    .numpy()
)

for cls_id, value in zip(
    classes95,
    values95,
):
    map95_by_class[
        int(cls_id)
    ] = (
        float(value)
        if value >= 0
        else np.nan
    )

classes50 = (
    map50_result["classes"]
    .cpu()
    .numpy()
    .astype(int)
)

values50 = (
    map50_result["map_per_class"]
    .cpu()
    .numpy()
)

for cls_id, value in zip(
    classes50,
    values50,
):
    map50_by_class[
        int(cls_id)
    ] = (
        float(value)
        if value >= 0
        else np.nan
    )


class_image_counts = defaultdict(
    set
)

class_instance_counts = defaultdict(
    int
)

for image_idx, target in enumerate(
    test_targets
):
    for cls_id in target[
        "labels"
    ].tolist():
        class_image_counts[
            int(cls_id)
        ].add(image_idx)

        class_instance_counts[
            int(cls_id)
        ] += 1


rows = []

for cls_id, class_name in enumerate(
    CLASS_NAMES
):
    stats = per_class_counts[
        cls_id
    ]

    tp = stats["tp"]
    fp = stats["fp"]
    fn = stats["fn"]

    p = (
        tp / (tp + fp)
        if tp + fp
        else np.nan
    )

    r = (
        tp / (tp + fn)
        if tp + fn
        else np.nan
    )

    f1 = (
        2 * p * r / (p + r)
        if (
            np.isfinite(p)
            and np.isfinite(r)
            and p + r > 0
        )
        else np.nan
    )

    cls_iou = (
        float(
            np.mean(
                stats["ious"]
            )
        )
        if stats["ious"]
        else np.nan
    )

    cls_dice = (
        2
        * cls_iou
        / (1 + cls_iou)
        if np.isfinite(
            cls_iou
        )
        else np.nan
    )

    rows.append({
        "Class": class_name,
        "Images":
            len(
                class_image_counts[
                    cls_id
                ]
            ),
        "Instances":
            class_instance_counts[
                cls_id
            ],
        "Box-P": p,
        "Box-R": r,
        "Box-F1": f1,
        "mAP50":
            map50_by_class.get(
                cls_id,
                np.nan,
            ),
        "mAP50-95":
            map95_by_class.get(
                cls_id,
                np.nan,
            ),
        "Mean Box IoU":
            cls_iou,
        "Mean Box Dice":
            cls_dice,
    })


per_class_df = pd.DataFrame(
    rows
).round(4)

per_class_df.to_csv(
    METRICS_DIR
    / "final_per_class_summary.csv",
    index=False,
)

display(per_class_df)
````

## 22. Raw and normalized confusion matrices

````python
N = NUM_CLASSES
BG = N

confusion = np.zeros(
    (N + 1, N + 1),
    dtype=np.int64,
)

for pred, gt in zip(
    test_predictions,
    test_targets,
):
    keep = (
        pred["scores"]
        >= CONF_THRESHOLD
    )

    p_boxes = pred["boxes"][keep]
    p_labels = pred["labels"][keep]

    g_boxes = gt["boxes"]
    g_labels = gt["labels"]

    used_p = set()
    used_g = set()

    if (
        len(p_boxes)
        and len(g_boxes)
    ):
        ious = box_iou(
            p_boxes,
            g_boxes,
        )

        candidates = []

        for pi in range(
            len(p_boxes)
        ):
            for gi in range(
                len(g_boxes)
            ):
                iou = float(
                    ious[pi, gi]
                )

                if (
                    iou
                    >= MATCH_IOU_THRESHOLD
                ):
                    candidates.append(
                        (
                            iou,
                            pi,
                            gi,
                        )
                    )

        candidates.sort(
            reverse=True
        )

        for (
            iou,
            pi,
            gi,
        ) in candidates:
            if (
                pi in used_p
                or gi in used_g
            ):
                continue

            used_p.add(pi)
            used_g.add(gi)

            confusion[
                int(g_labels[gi]),
                int(p_labels[pi]),
            ] += 1

    for gi, cls_id in enumerate(
        g_labels.tolist()
    ):
        if gi not in used_g:
            confusion[
                int(cls_id),
                BG,
            ] += 1

    for pi, cls_id in enumerate(
        p_labels.tolist()
    ):
        if pi not in used_p:
            confusion[
                BG,
                int(cls_id),
            ] += 1


labels_cm = (
    CLASS_NAMES
    + ["background"]
)

row_sums = confusion.sum(
    axis=1,
    keepdims=True,
)

normalized = np.divide(
    confusion,
    row_sums,
    out=np.zeros_like(
        confusion,
        dtype=float,
    ),
    where=row_sums != 0,
)

pd.DataFrame(
    confusion,
    index=labels_cm,
    columns=labels_cm,
).to_csv(
    METRICS_DIR
    / "confusion_matrix.csv"
)

pd.DataFrame(
    normalized,
    index=labels_cm,
    columns=labels_cm,
).to_csv(
    METRICS_DIR
    / "confusion_matrix_normalized.csv"
)


def plot_confusion(
    matrix,
    title,
    filename,
    fmt,
):
    fig, ax = plt.subplots(
        figsize=(11, 9)
    )

    ax.imshow(matrix)

    ax.set_xticks(
        range(len(labels_cm))
    )

    ax.set_yticks(
        range(len(labels_cm))
    )

    ax.set_xticklabels(
        labels_cm,
        rotation=45,
        ha="right",
    )

    ax.set_yticklabels(
        labels_cm
    )

    ax.set_xlabel(
        "Predicted class"
    )

    ax.set_ylabel(
        "Ground-truth class"
    )

    ax.set_title(title)

    for i in range(
        matrix.shape[0]
    ):
        for j in range(
            matrix.shape[1]
        ):
            value = matrix[i, j]

            text = (
                f"{value:.2f}"
                if fmt == "float"
                else str(int(value))
            )

            if value != 0:
                ax.text(
                    j,
                    i,
                    text,
                    ha="center",
                    va="center",
                    fontsize=7,
                )

    plt.tight_layout()

    plt.savefig(
        PLOTS_DIR / filename,
        dpi=300,
        bbox_inches="tight",
    )

    plt.show()


plot_confusion(
    confusion,
    "PVTv2-B0 + RetinaNet Confusion Matrix",
    "confusion_matrix.png",
    "int",
)

plot_confusion(
    normalized,
    "PVTv2-B0 + RetinaNet Normalized Confusion Matrix",
    "confusion_matrix_normalized.png",
    "float",
)
````

## 23. Ground Truth vs Prediction

````python
from matplotlib.patches import Rectangle

def show_box(
    ax,
    box,
    label,
    linewidth=1.5,
):
    x1, y1, x2, y2 = box

    rect = Rectangle(
        (x1, y1),
        x2 - x1,
        y2 - y1,
        fill=False,
        linewidth=linewidth,
    )

    ax.add_patch(rect)

    ax.text(
        x1,
        max(10, y1),
        label,
        fontsize=7,
        bbox=dict(
            facecolor="white",
            alpha=0.65,
            edgecolor="none",
        ),
    )


defect_indices = [
    i
    for i, target
    in enumerate(test_targets)
    if len(target["boxes"]) > 0
]

random.seed(SEED)

sample_indices = random.sample(
    defect_indices,
    min(
        6,
        len(defect_indices),
    ),
)

fig, axes = plt.subplots(
    len(sample_indices),
    2,
    figsize=(
        12,
        4 * len(sample_indices),
    ),
)

if len(sample_indices) == 1:
    axes = np.array([axes])

for row, idx in enumerate(
    sample_indices
):
    image_tensor, gt, _ = (
        test_dataset[idx]
    )

    image_np = (
        image_tensor
        .permute(1, 2, 0)
        .numpy()
        .clip(0, 1)
    )

    pred = test_predictions[idx]

    axes[row, 0].imshow(
        image_np
    )

    for box, cls_id in zip(
        gt["boxes"].numpy(),
        gt["labels"].numpy(),
    ):
        show_box(
            axes[row, 0],
            box,
            CLASS_NAMES[
                int(cls_id)
            ],
        )

    axes[row, 0].set_title(
        "Ground Truth"
    )

    axes[row, 0].axis("off")

    axes[row, 1].imshow(
        image_np
    )

    keep = (
        pred["scores"]
        >= CONF_THRESHOLD
    )

    for (
        box,
        cls_id,
        score,
    ) in zip(
        pred["boxes"][keep].numpy(),
        pred["labels"][keep].numpy(),
        pred["scores"][keep].numpy(),
    ):
        show_box(
            axes[row, 1],
            box,
            (
                f"{CLASS_NAMES[int(cls_id)]} "
                f"{score:.2f}"
            ),
        )

    axes[row, 1].set_title(
        "PVTv2-B0 + RetinaNet Prediction"
    )

    axes[row, 1].axis("off")

plt.tight_layout()

plt.savefig(
    PLOTS_DIR
    / "ground_truth_vs_prediction.png",
    dpi=300,
    bbox_inches="tight",
)

plt.show()
````
