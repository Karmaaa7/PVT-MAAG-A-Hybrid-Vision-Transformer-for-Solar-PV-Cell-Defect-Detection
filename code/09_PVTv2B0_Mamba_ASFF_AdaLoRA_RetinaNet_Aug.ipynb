# CSE465 — PVTv2-B0 + Mamba + ASFF + AdaLoRA + RetinaNet (Augmented)

This is the **Notebook 09 canonical AdaLoRA experiment**.

It starts from the trained A2 checkpoint:

`07_PVTv2B0_Mamba_ASFF_RetinaNet_Aug/weights/best.pt`

and adapts:

`PVTv2-B0 → deep bidirectional Mamba → FPN → ASFF → RetinaNet`

using **AdaLoRA inside the PVTv2 backbone**.

The dataset split, 640 × 640 input, augmentation, Mamba, ASFF, RetinaNet,
effective batch size, seed, confidence threshold, IoU convention and final
evaluation code are kept the same as A2.

## Comparison logic

- **A2**: PVTv2-B0 + deep bidirectional Mamba + FPN + ASFF + RetinaNet
- **Notebook 09**: A2 + canonical AdaLoRA fine-tuning
- Notebook 09 starts directly from A2 `best.pt`
- Notebook 08 is not used as the source checkpoint

## Canonical AdaLoRA used here

This notebook uses Hugging Face PEFT AdaLoRA.

- SVD-style low-rank updates are inserted into PVTv2 attention and MLP linear layers.
- Initial rank: `12`
- Target average rank: `8`
- Rank allocation uses importance scores during training.
- `update_and_allocate()` is called after every optimizer update.
- Initial no-pruning phase: first `2` epochs.
- Final fixed-rank fine-tuning phase: last `5` epochs.
- All transferred A2 parameters remain frozen.
- Only AdaLoRA parameters are optimized.
- Epoch 0 is the exact transferred A2 model.
- The trained checkpoint replaces the A2 fallback only after a validation
  mAP@50–95 improvement greater than the configured baseline margin.

## Training

- source checkpoint: A2 `best.pt`
- image size: 640 × 640
- physical batch: 8
- gradient accumulation: 2
- effective batch: 16
- maximum epochs: 25
- patience: 5
- best model selected by validation mAP@50–95

## 9. Dataset class — same augmentation as A0/A1/A2

No augmentation variable is changed for the AdaLoRA experiment.


```python
IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png",
    ".bmp", ".tif", ".tiff",
}


def letterbox_image(
    image,
    boxes,
    size=640,
    fill=114,
):
    h, w = image.shape[:2]

    scale = min(
        size / w,
        size / h,
    )

    new_w = int(
        round(w * scale)
    )

    new_h = int(
        round(h * scale)
    )

    resized = cv2.resize(
        image,
        (new_w, new_h),
        interpolation=cv2.INTER_LINEAR,
    )

    pad_x = (
        size - new_w
    ) // 2

    pad_y = (
        size - new_h
    ) // 2

    canvas = np.full(
        (size, size, 3),
        fill,
        dtype=np.uint8,
    )

    canvas[
        pad_y:pad_y + new_h,
        pad_x:pad_x + new_w,
    ] = resized

    boxes = np.asarray(
        boxes,
        dtype=np.float32,
    ).reshape(-1, 4)

    if len(boxes):
        boxes = boxes.copy()

        boxes[:, [0, 2]] = (
            boxes[:, [0, 2]]
            * scale
            + pad_x
        )

        boxes[:, [1, 3]] = (
            boxes[:, [1, 3]]
            * scale
            + pad_y
        )

    return canvas, boxes


train_transform = v2.Compose([
    v2.RandomApply(
        [
            v2.ColorJitter(
                brightness=AUGMENTATION["brightness"],
                contrast=0.0,
                saturation=0.0,
                hue=0.0,
            )
        ],
        p=AUGMENTATION["brightness_p"],
    ),

    v2.RandomApply(
        [
            v2.ColorJitter(
                brightness=0.0,
                contrast=AUGMENTATION["contrast"],
                saturation=0.0,
                hue=0.0,
            )
        ],
        p=AUGMENTATION["contrast_p"],
    ),

    v2.RandomApply(
        [
            v2.GaussianBlur(
                kernel_size=AUGMENTATION[
                    "gaussian_blur_kernel"
                ],
                sigma=(
                    AUGMENTATION[
                        "gaussian_blur_sigma_min"
                    ],
                    AUGMENTATION[
                        "gaussian_blur_sigma_max"
                    ],
                ),
            )
        ],
        p=AUGMENTATION["gaussian_blur_p"],
    ),

    v2.RandomHorizontalFlip(
        p=AUGMENTATION["fliplr"]
    ),

    v2.RandomVerticalFlip(
        p=AUGMENTATION["flipud"]
    ),
])


class PVDetectionDataset(Dataset):
    def __init__(
        self,
        dataset_dir,
        split,
        image_size=640,
        augment=False,
    ):
        self.image_dir = (
            Path(dataset_dir)
            / "images"
            / split
        )

        self.label_dir = (
            Path(dataset_dir)
            / "labels"
            / split
        )

        self.split = split
        self.image_size = image_size
        self.augment = augment

        self.images = sorted(
            p
            for p in self.image_dir.iterdir()
            if p.suffix.lower()
            in IMAGE_EXTENSIONS
        )

    def __len__(self):
        return len(self.images)

    def _load_raw(self, idx):
        image_path = self.images[idx]

        label_path = (
            self.label_dir
            / f"{image_path.stem}.txt"
        )

        image = cv2.imread(
            str(image_path)
        )

        if image is None:
            raise RuntimeError(
                f"Could not read: {image_path}"
            )

        image = cv2.cvtColor(
            image,
            cv2.COLOR_BGR2RGB,
        )

        h, w = image.shape[:2]

        boxes = []
        labels = []

        if label_path.exists():
            text = (
                label_path
                .read_text()
                .strip()
            )

            if text:
                for line in text.splitlines():
                    (
                        cls_id,
                        xc,
                        yc,
                        bw,
                        bh,
                    ) = map(
                        float,
                        line.split(),
                    )

                    x1 = (
                        xc - bw / 2
                    ) * w

                    y1 = (
                        yc - bh / 2
                    ) * h

                    x2 = (
                        xc + bw / 2
                    ) * w

                    y2 = (
                        yc + bh / 2
                    ) * h

                    boxes.append(
                        [x1, y1, x2, y2]
                    )

                    labels.append(
                        int(cls_id)
                    )

        boxes = np.asarray(
            boxes,
            dtype=np.float32,
        ).reshape(-1, 4)

        labels = np.asarray(
            labels,
            dtype=np.int64,
        ).reshape(-1)

        return (
            image,
            boxes,
            labels,
            str(image_path),
        )

    def __getitem__(self, idx):
        (
            image,
            boxes,
            labels,
            image_path,
        ) = self._load_raw(idx)

        image, boxes = letterbox_image(
            image,
            boxes,
            size=self.image_size,
        )

        image = tv_tensors.Image(
            torch.from_numpy(
                np.ascontiguousarray(
                    image
                )
            ).permute(2, 0, 1)
        )

        boxes = tv_tensors.BoundingBoxes(
            torch.as_tensor(
                boxes,
                dtype=torch.float32,
            ).reshape(-1, 4),
            format="XYXY",
            canvas_size=(
                self.image_size,
                self.image_size,
            ),
        )

        target = {
            "boxes": boxes,
            "labels": torch.as_tensor(
                labels,
                dtype=torch.int64,
            ),
        }

        original_box_count = len(
            target["boxes"]
        )

        if self.augment:
            image, target = train_transform(
                image,
                target,
            )

        image = (
            image
            .as_subclass(torch.Tensor)
            .to(torch.float32)
            / 255.0
        )

        if (
            self.augment
            and random.random()
            < AUGMENTATION["gaussian_noise_p"]
        ):
            image = torch.clamp(
                image
                + torch.randn_like(image)
                * AUGMENTATION[
                    "gaussian_noise_std"
                ],
                0.0,
                1.0,
            )

        boxes_out = (
            target["boxes"]
            .as_subclass(torch.Tensor)
            .to(torch.float32)
        )

        labels_out = (
            target["labels"]
            .to(torch.int64)
        )

        if len(boxes_out) != original_box_count:
            raise RuntimeError(
                f"Unexpected box-count change: "
                f"{image_path}"
            )

        if len(boxes_out):
            if not torch.isfinite(
                boxes_out
            ).all():
                raise RuntimeError(
                    f"Non-finite box: "
                    f"{image_path}"
                )

            if not (
                (
                    boxes_out[:, 2]
                    > boxes_out[:, 0]
                ).all()
                and (
                    boxes_out[:, 3]
                    > boxes_out[:, 1]
                ).all()
            ):
                raise RuntimeError(
                    f"Invalid box: "
                    f"{image_path}"
                )

        target_out = {
            "boxes": boxes_out,
            "labels": labels_out,
            "image_id": torch.tensor(
                [idx],
                dtype=torch.int64,
            ),
        }

        return (
            image,
            target_out,
            image_path,
        )


def collate_fn(batch):
    images, targets, paths = zip(*batch)

    return (
        list(images),
        list(targets),
        list(paths),
    )


def seed_worker(worker_id):
    worker_seed = (
        torch.initial_seed()
        % 2**32
    )

    np.random.seed(worker_seed)
    random.seed(worker_seed)


loader_generator = torch.Generator()
loader_generator.manual_seed(SEED)


train_dataset = PVDetectionDataset(
    DATASET_DIR,
    "train",
    IMAGE_SIZE,
    augment=True,
)

val_dataset = PVDetectionDataset(
    DATASET_DIR,
    "val",
    IMAGE_SIZE,
    augment=False,
)

test_dataset = PVDetectionDataset(
    DATASET_DIR,
    "test",
    IMAGE_SIZE,
    augment=False,
)


train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    collate_fn=collate_fn,
    worker_init_fn=seed_worker,
    generator=loader_generator,
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    collate_fn=collate_fn,
    worker_init_fn=seed_worker,
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True,
    collate_fn=collate_fn,
    worker_init_fn=seed_worker,
)

print(
    "Train:", len(train_dataset),
    "(augmentation ON)"
)

print(
    "Val  :", len(val_dataset),
    "(augmentation OFF)"
)

print(
    "Test :", len(test_dataset),
    "(augmentation OFF)"
)

print(
    "Effective batch:",
    EFFECTIVE_BATCH
)
```

### 8.1 Mandatory augmentation preflight

This configuration does not crop or geometrically deform boxes, so the
preflight requires **100% box-count retention**.

It checks 1,000 random training samples before the long run.

```python
from matplotlib.patches import Rectangle

def augmentation_preflight(
    dataset,
    sample_count=1000,
):
    selector = random.Random(SEED)

    indices = [
        selector.randrange(
            len(dataset)
        )
        for _ in range(sample_count)
    ]

    raw_counts = np.zeros(
        NUM_CLASSES,
        dtype=np.int64,
    )

    aug_counts = np.zeros(
        NUM_CLASSES,
        dtype=np.int64,
    )

    invalid = 0

    for idx in indices:
        (
            _,
            raw_boxes,
            raw_labels,
            _,
        ) = dataset._load_raw(idx)

        raw_counts += np.bincount(
            raw_labels,
            minlength=NUM_CLASSES,
        )

        image, target, _ = dataset[idx]

        boxes = (
            target["boxes"]
            .cpu()
            .numpy()
        )

        labels = (
            target["labels"]
            .cpu()
            .numpy()
        )

        aug_counts += np.bincount(
            labels,
            minlength=NUM_CLASSES,
        )

        if len(boxes) != len(raw_boxes):
            invalid += 1
            continue

        if len(boxes):
            if not np.isfinite(
                boxes
            ).all():
                invalid += 1
                continue

            if not (
                (
                    boxes[:, 2]
                    > boxes[:, 0]
                ).all()
                and (
                    boxes[:, 3]
                    > boxes[:, 1]
                ).all()
            ):
                invalid += 1

    print("=" * 70)
    print("AUGMENTATION PREFLIGHT")
    print("=" * 70)
    print(
        "Samples checked:",
        sample_count,
    )
    print(
        "Invalid samples:",
        invalid,
    )

    if invalid != 0:
        raise RuntimeError(
            "Augmentation preflight failed. "
            "Do NOT start training."
        )

    print(
        "\nClass-wise annotation retention:"
    )

    for cls_id, name in enumerate(
        CLASS_NAMES
    ):
        raw_n = int(
            raw_counts[cls_id]
        )

        aug_n = int(
            aug_counts[cls_id]
        )

        if raw_n == 0:
            print(
                f"{name:25s}: "
                "not sampled"
            )
            continue

        print(
            f"{name:25s}: "
            f"{aug_n}/{raw_n} "
            f"({aug_n/raw_n:.1%})"
        )

        if aug_n != raw_n:
            raise RuntimeError(
                f"Annotation count changed "
                f"for {name}. "
                "Do NOT train."
            )

    print(
        "\nPreflight PASSED."
    )


augmentation_preflight(
    train_dataset,
    sample_count=1000,
)


fig, axes = plt.subplots(
    2,
    4,
    figsize=(20, 10),
)

axes = axes.ravel()

preview_rng = random.Random(
    SEED + 1
)

for ax in axes:
    idx = preview_rng.randrange(
        len(train_dataset)
    )

    image, target, _ = (
        train_dataset[idx]
    )

    ax.imshow(
        image
        .permute(1, 2, 0)
        .cpu()
        .numpy()
        .clip(0, 1)
    )

    for box, cls_id in zip(
        target["boxes"].numpy(),
        target["labels"].numpy(),
    ):
        x1, y1, x2, y2 = box

        ax.add_patch(
            Rectangle(
                (x1, y1),
                x2 - x1,
                y2 - y1,
                fill=False,
                linewidth=1.5,
            )
        )

        ax.text(
            x1,
            max(10, y1),
            CLASS_NAMES[
                int(cls_id)
            ],
            fontsize=7,
            bbox=dict(
                facecolor="white",
                alpha=0.65,
                edgecolor="none",
            ),
        )

    ax.axis("off")

plt.suptitle(
    "PVTv2-B0 final augmentation preview",
    fontsize=14,
)

plt.tight_layout()
plt.show()

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

print(
    "RNGs reset after preflight."
)
```

## 10. Build A2 architecture, load A2 best.pt, then inject canonical AdaLoRA

```python
class DeepBidirectionalMamba(
    nn.Module
):
    def __init__(
        self,
        channels,
        d_state=16,
        d_conv=4,
        expand=2,
        gate_init=0.0,
    ):
        super().__init__()

        self.channels = channels

        self.norm = nn.LayerNorm(
            channels
        )

        self.mamba = Mamba(
            d_model=channels,
            d_state=d_state,
            d_conv=d_conv,
            expand=expand,
        )

        self.gate = nn.Parameter(
            torch.tensor(
                float(gate_init)
            )
        )

    def forward(
        self,
        x,
    ):
        # x: [B, C, H, W]
        b, c, h, w = x.shape

        if c != self.channels:
            raise RuntimeError(
                f"Mamba expected {self.channels} channels, "
                f"received {c}."
            )

        seq = (
            x
            .flatten(2)
            .transpose(1, 2)
            .contiguous()
        )

        z = self.norm(
            seq
        )

        y_forward = self.mamba(
            z
        )

        z_reverse = torch.flip(
            z,
            dims=[1],
        )

        y_reverse = self.mamba(
            z_reverse
        )

        y_reverse = torch.flip(
            y_reverse,
            dims=[1],
        )

        y = (
            y_forward
            + y_reverse
        ) * 0.5

        seq = (
            seq
            + torch.tanh(
                self.gate
            ) * y
        )

        out = (
            seq
            .transpose(1, 2)
            .reshape(
                b,
                c,
                h,
                w,
            )
            .contiguous()
        )

        return out


class ASFFFusionLevel(
    nn.Module
):
    def __init__(
        self,
        channels=256,
        compress_channels=8,
        num_levels=3,
        gate_init=0.1,
    ):
        super().__init__()

        self.channels = channels
        self.num_levels = num_levels

        self.compress = nn.ModuleList([
            nn.Sequential(
                nn.Conv2d(
                    channels,
                    compress_channels,
                    kernel_size=1,
                    bias=True,
                ),
                nn.ReLU(
                    inplace=True
                ),
            )
            for _ in range(
                num_levels
            )
        ])

        self.weight_logits = nn.Conv2d(
            compress_channels
            * num_levels,
            num_levels,
            kernel_size=1,
            bias=True,
        )

        # Near-uniform softmax weights, but avoid an all-zero
        # kernel so the weighting branch can differentiate levels.
        nn.init.normal_(
            self.weight_logits.weight,
            mean=0.0,
            std=0.01,
        )
        nn.init.zeros_(
            self.weight_logits.bias
        )

        self.refine = nn.Conv2d(
            channels,
            channels,
            kernel_size=3,
            padding=1,
            bias=True,
        )

        # A small non-zero gate lets gradients reach ASFF
        # immediately without overwhelming the transferred A1 path.
        self.gate = nn.Parameter(
            torch.tensor(
                float(gate_init)
            )
        )

    def forward(
        self,
        aligned_features,
        target_index,
    ):
        if len(
            aligned_features
        ) != self.num_levels:
            raise RuntimeError(
                "ASFF received the wrong "
                "number of feature levels."
            )

        compressed = [
            layer(feature)
            for layer, feature
            in zip(
                self.compress,
                aligned_features,
            )
        ]

        logits = self.weight_logits(
            torch.cat(
                compressed,
                dim=1,
            )
        )

        # [B, 3, H, W]
        weights = torch.softmax(
            logits,
            dim=1,
        )

        fused = torch.zeros_like(
            aligned_features[
                target_index
            ]
        )

        for level_idx, feature in enumerate(
            aligned_features
        ):
            fused = (
                fused
                + weights[
                    :,
                    level_idx:
                    level_idx + 1,
                ]
                * feature
            )

        fused = self.refine(
            fused
        )

        base = aligned_features[
            target_index
        ]

        out = (
            base
            + torch.tanh(
                self.gate
            ) * fused
        )

        return out


class AdaptiveSpatialFeatureFusion(
    nn.Module
):
    def __init__(
        self,
        channels=256,
        compress_channels=8,
        num_levels=3,
        gate_init=0.1,
    ):
        super().__init__()

        self.channels = channels
        self.num_levels = num_levels

        self.levels = nn.ModuleList([
            ASFFFusionLevel(
                channels=channels,
                compress_channels=
                    compress_channels,
                num_levels=num_levels,
                gate_init=gate_init,
            )
            for _ in range(
                num_levels
            )
        ])

    @staticmethod
    def _resize_to(
        feature,
        target_hw,
    ):
        current_hw = (
            feature.shape[-2],
            feature.shape[-1],
        )

        if current_hw == target_hw:
            return feature

        # Downsample with adaptive average pooling.
        if (
            current_hw[0]
            >= target_hw[0]
            and current_hw[1]
            >= target_hw[1]
        ):
            return F.adaptive_avg_pool2d(
                feature,
                output_size=target_hw,
            )

        # Upsample with bilinear interpolation.
        return F.interpolate(
            feature,
            size=target_hw,
            mode="bilinear",
            align_corners=False,
        )

    def forward(
        self,
        features,
    ):
        if not isinstance(
            features,
            OrderedDict,
        ):
            features = OrderedDict(
                features
            )

        keys = list(
            features.keys()
        )

        if len(keys) < self.num_levels:
            raise RuntimeError(
                "ASFF requires at least "
                f"{self.num_levels} FPN levels."
            )

        # Main RetinaNet pyramid levels: P3/P4/P5.
        main_keys = keys[
            :self.num_levels
        ]

        main_features = [
            features[key]
            for key in main_keys
        ]

        for feature in main_features:
            if feature.shape[1] != self.channels:
                raise RuntimeError(
                    "ASFF expected "
                    f"{self.channels} channels, "
                    f"received {feature.shape[1]}."
                )

        output = OrderedDict(
            features
        )

        for (
            target_index,
            target_key,
        ) in enumerate(
            main_keys
        ):
            target_hw = (
                main_features[
                    target_index
                ].shape[-2:]
            )

            aligned = [
                self._resize_to(
                    feature,
                    target_hw,
                )
                for feature
                in main_features
            ]

            output[
                target_key
            ] = self.levels[
                target_index
            ](
                aligned,
                target_index,
            )

        # Remaining FPN levels (P6/P7) are unchanged.
        return output


class PVTv2B0MambaASFFFPN(
    nn.Module
):
    def __init__(
        self,
        pretrained=False,
        out_channels=256,
    ):
        super().__init__()

        self.body = timm.create_model(
            "pvt_v2_b0",
            pretrained=pretrained,
            features_only=True,
            out_indices=(1, 2, 3),
        )

        feature_channels = (
            self.body
            .feature_info
            .channels()
        )

        feature_reductions = (
            self.body
            .feature_info
            .reduction()
        )

        print(
            "PVTv2 feature channels:",
            feature_channels,
        )

        print(
            "PVTv2 feature strides :",
            feature_reductions,
        )

        deepest_channels = (
            feature_channels[-1]
        )

        if deepest_channels != 256:
            raise RuntimeError(
                "Expected deepest PVTv2-B0 selected "
                f"feature to have 256 channels, got "
                f"{deepest_channels}."
            )

        self.deep_mamba = (
            DeepBidirectionalMamba(
                channels=deepest_channels,
                d_state=16,
                d_conv=4,
                expand=2,
                # The actual trained A1 gate value is loaded
                # from A1 best.pt.
                gate_init=0.0,
            )
        )

        self.fpn = (
            FeaturePyramidNetwork(
                in_channels_list=
                    feature_channels,

                out_channels=
                    out_channels,

                extra_blocks=
                    LastLevelP6P7(
                        feature_channels[-1],
                        out_channels,
                    ),
            )
        )

        self.asff = (
            AdaptiveSpatialFeatureFusion(
                channels=ASFF_CONFIG[
                    "channels"
                ],
                compress_channels=
                    ASFF_CONFIG[
                        "compress_channels"
                    ],
                num_levels=3,
                gate_init=
                    ASFF_CONFIG[
                        "residual_gate_init"
                    ],
            )
        )

        self.out_channels = (
            out_channels
        )

    def forward(
        self,
        x,
    ):
        features = self.body(
            x
        )

        # Existing A1 Mamba path.
        features[-1] = (
            self.deep_mamba(
                features[-1]
            )
        )

        features = OrderedDict(
            (
                str(i),
                feat,
            )
            for i, feat
            in enumerate(features)
        )

        fpn_features = self.fpn(
            features
        )

        # Existing A2 ASFF module.
        return self.asff(
            fpn_features
        )


def build_model():
    backbone = (
        PVTv2B0MambaASFFFPN(
            pretrained=False,
            out_channels=256,
        )
    )

    model = RetinaNet(
        backbone=backbone,
        num_classes=NUM_CLASSES,
        min_size=IMAGE_SIZE,
        max_size=IMAGE_SIZE,
        image_mean=[
            0.485,
            0.456,
            0.406,
        ],
        image_std=[
            0.229,
            0.224,
            0.225,
        ],
        score_thresh=0.001,
        detections_per_img=100,
    )

    return model


model = build_model()


# ------------------------------------------------------------
# Load the trained A2 detector BEFORE injecting LoRA.
# ------------------------------------------------------------
a2_checkpoint = torch.load(
    A2_BEST_PT,
    map_location="cpu",
    weights_only=False,
)

a2_state = a2_checkpoint["model_state"]

model.load_state_dict(
    a2_state,
    strict=True,
)

print(
    f"Loaded A2 checkpoint from epoch "
    f"{a2_checkpoint.get('epoch', 'unknown')}"
)

print(
    "A2 validation mAP50-95 stored in checkpoint:",
    a2_checkpoint.get(
        "best_map50_95",
        "not stored",
    ),
)



# ------------------------------------------------------------
# Resolve canonical AdaLoRA schedule in OPTIMIZER steps.
# ------------------------------------------------------------
UPDATES_PER_EPOCH = math.ceil(
    len(train_loader)
    / ACCUM_STEPS
)

TOTAL_UPDATE_STEPS = (
    MAX_EPOCHS
    * UPDATES_PER_EPOCH
)

ADALORA_TINIT_STEPS = (
    ADALORA_INIT_EPOCHS
    * UPDATES_PER_EPOCH
)

ADALORA_TFINAL_STEPS = (
    ADALORA_FINAL_EPOCHS
    * UPDATES_PER_EPOCH
)

if (
    ADALORA_TINIT_STEPS
    >= TOTAL_UPDATE_STEPS
    - ADALORA_TFINAL_STEPS
):
    raise RuntimeError(
        "AdaLoRA schedule leaves no budgeting phase."
    )

training_config["adalora"][
    "updates_per_epoch"
] = UPDATES_PER_EPOCH

training_config["adalora"][
    "total_update_steps"
] = TOTAL_UPDATE_STEPS

training_config["adalora"][
    "tinit_steps"
] = ADALORA_TINIT_STEPS

training_config["adalora"][
    "tfinal_steps"
] = ADALORA_TFINAL_STEPS

with open(
    CONFIG_DIR
    / "training_config.yaml",
    "w",
) as f:
    yaml.safe_dump(
        training_config,
        f,
        sort_keys=False,
    )


# ------------------------------------------------------------
# Inject canonical AdaLoRA into the PVTv2 body only.
# PEFT supports custom nn.Module models when targets are specified.
# ------------------------------------------------------------
peft_adalora_config = AdaLoraConfig(
    init_r=ADALORA_INIT_R,
    target_r=ADALORA_TARGET_R,

    lora_alpha=ADALORA_ALPHA,
    lora_dropout=ADALORA_DROPOUT,

    target_modules=ADALORA_CONFIG[
        "target_modules"
    ],

    tinit=ADALORA_TINIT_STEPS,
    tfinal=ADALORA_TFINAL_STEPS,
    deltaT=ADALORA_DELTA_T,

    beta1=ADALORA_BETA1,
    beta2=ADALORA_BETA2,

    orth_reg_weight=ADALORA_ORTH_REG_WEIGHT,

    total_step=TOTAL_UPDATE_STEPS,

    bias="none",
    task_type=None,
)

model.backbone.body = get_peft_model(
    model.backbone.body,
    peft_adalora_config,
)

adalora_body = model.backbone.body


# ------------------------------------------------------------
# Keep every transferred A2 component outside the PVT PEFT body frozen.
# get_peft_model already freezes the PVT base weights.
# ------------------------------------------------------------
for parameter in (
    model.backbone.deep_mamba.parameters()
):
    parameter.requires_grad = False

for parameter in (
    model.backbone.fpn.parameters()
):
    parameter.requires_grad = False

for parameter in (
    model.backbone.asff.parameters()
):
    parameter.requires_grad = False

for parameter in (
    model.head.parameters()
):
    parameter.requires_grad = False


# ------------------------------------------------------------
# Verify only AdaLoRA parameters are trainable.
# ------------------------------------------------------------
trainable_named = [
    (name, parameter)
    for name, parameter
    in model.named_parameters()
    if parameter.requires_grad
]

if not trainable_named:
    raise RuntimeError(
        "AdaLoRA injection produced zero trainable parameters."
    )

unexpected_trainable = [
    name
    for name, _
    in trainable_named
    if "lora_" not in name
]

if unexpected_trainable:
    raise RuntimeError(
        "Unexpected non-AdaLoRA parameters are trainable: "
        + ", ".join(
            unexpected_trainable[:10]
        )
    )

adalora_params_list = [
    parameter
    for _, parameter
    in trainable_named
]

adalora_params_count = sum(
    parameter.numel()
    for parameter
    in adalora_params_list
)

targeted_module_names = list(
    getattr(
        adalora_body,
        "targeted_module_names",
        [],
    )
)

if not targeted_module_names:
    raise RuntimeError(
        "PEFT reported zero targeted AdaLoRA modules."
    )


# ------------------------------------------------------------
# AdaLoRA rank/allocator helpers.
# ------------------------------------------------------------
def get_adalora_tuner(peft_body):
    candidates = [
        peft_body,
        getattr(
            peft_body,
            "base_model",
            None,
        ),
    ]

    for candidate in candidates:
        if (
            candidate is not None
            and hasattr(
                candidate,
                "update_and_allocate",
            )
        ):
            return candidate

    raise RuntimeError(
        "AdaLoRA update_and_allocate() is unavailable. "
        "Do not train because dynamic rank allocation would be disabled."
    )


adalora_tuner = get_adalora_tuner(
    adalora_body
)


def get_rank_allocator(peft_body):
    candidates = [
        peft_body,
        getattr(
            peft_body,
            "base_model",
            None,
        ),
        adalora_tuner,
    ]

    for candidate in candidates:
        allocator = getattr(
            candidate,
            "rankallocator",
            None,
        )

        if allocator is not None:
            return allocator

    return None


def current_rank_pattern():
    config = (
        adalora_body
        .peft_config[
            "default"
        ]
    )

    pattern = getattr(
        config,
        "rank_pattern",
        None,
    )

    if pattern is None:
        return None

    return {
        str(k): int(v)
        for k, v
        in pattern.items()
    }


def adalora_rank_stats():
    ranks = []

    for name, parameter in (
        adalora_body.named_parameters()
    ):
        if "lora_E" not in name:
            continue

        rank = int(
            torch.count_nonzero(
                parameter.detach()
            ).item()
        )

        ranks.append(rank)

    if not ranks:
        return {
            "modules": 0,
            "active_rank_total": 0,
            "active_rank_mean": np.nan,
            "active_rank_min": np.nan,
            "active_rank_max": np.nan,
        }

    return {
        "modules":
            len(ranks),

        "active_rank_total":
            int(np.sum(ranks)),

        "active_rank_mean":
            float(np.mean(ranks)),

        "active_rank_min":
            int(np.min(ranks)),

        "active_rank_max":
            int(np.max(ranks)),
    }


def adalora_phase(global_step):
    if global_step <= ADALORA_TINIT_STEPS:
        return "AdaLoRA initialization"

    if (
        global_step
        <= TOTAL_UPDATE_STEPS
        - ADALORA_TFINAL_STEPS
    ):
        return "AdaLoRA budget allocation"

    return "AdaLoRA final fixed-rank fine-tuning"


def adalora_orthogonal_regularization(
    peft_body,
    weight,
):
    regu_loss = None
    count = 0

    for name, parameter in (
        peft_body.named_parameters()
    ):
        if (
            "lora_A" not in name
            and "lora_B" not in name
        ):
            continue

        if parameter.ndim != 2:
            continue

        if "lora_A" in name:
            covariance = (
                parameter
                @ parameter.T
            )
        else:
            covariance = (
                parameter.T
                @ parameter
            )

        identity = torch.eye(
            covariance.shape[0],
            device=covariance.device,
            dtype=covariance.dtype,
        )

        term = torch.norm(
            covariance - identity,
            p="fro",
        )

        regu_loss = (
            term
            if regu_loss is None
            else regu_loss + term
        )

        count += 1

    if count == 0:
        return torch.zeros(
            (),
            device=DEVICE,
        )

    return (
        float(weight)
        * regu_loss
        / count
    )


def capture_rankallocator_state():
    allocator = get_rank_allocator(
        adalora_body
    )

    if allocator is None:
        return None

    state = {}

    for attr in [
        "ipt",
        "exp_avg_ipt",
        "exp_avg_unc",
    ]:
        value = getattr(
            allocator,
            attr,
            None,
        )

        if isinstance(
            value,
            dict,
        ):
            state[attr] = {
                str(k):
                    v.detach().cpu()
                for k, v
                in value.items()
            }

    return state


def restore_rankallocator_state(
    saved_state,
):
    if not saved_state:
        return

    allocator = get_rank_allocator(
        adalora_body
    )

    if allocator is None:
        warnings.warn(
            "RankAllocator state could not be restored."
        )
        return

    for attr, value in (
        saved_state.items()
    ):
        if not isinstance(
            value,
            dict,
        ):
            continue

        setattr(
            allocator,
            attr,
            {
                str(k):
                    v.to(DEVICE)
                for k, v
                in value.items()
            },
        )


# AdaLoRA uses zero-initialized singular values.
# This must keep epoch 0 exactly equal to A2.
initial_nonzero_E = 0

for name, parameter in (
    adalora_body.named_parameters()
):
    if "lora_E" in name:
        initial_nonzero_E += int(
            torch.count_nonzero(
                parameter.detach()
            ).item()
        )

if initial_nonzero_E != 0:
    raise RuntimeError(
        "AdaLoRA lora_E must be zero before training. "
        "Epoch 0 would not match A2."
    )


model = model.to(DEVICE)

total_params = sum(
    p.numel()
    for p in model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

mamba_params = sum(
    p.numel()
    for p
    in model.backbone
    .deep_mamba
    .parameters()
)

asff_params = sum(
    p.numel()
    for p
    in model.backbone
    .asff
    .parameters()
)

print(
    "PEFT version:",
    peft.__version__,
)

print(
    "AdaLoRA targeted modules:",
    len(
        targeted_module_names
    ),
)

print(
    "AdaLoRA trainable parameters:",
    f"{adalora_params_count:,}",
    f"({adalora_params_count/1e6:.4f} M)",
)

print(
    "Total parameters:",
    f"{total_params:,}",
    f"({total_params/1e6:.3f} M)",
)

print(
    "Trainable parameters:",
    f"{trainable_params:,}",
    f"({trainable_params/1e6:.4f} M)",
)

print(
    "AdaLoRA optimizer updates/epoch:",
    UPDATES_PER_EPOCH,
)

print(
    "AdaLoRA total optimizer updates:",
    TOTAL_UPDATE_STEPS,
)

print(
    "AdaLoRA tinit/tfinal:",
    ADALORA_TINIT_STEPS,
    ADALORA_TFINAL_STEPS,
)

print(
    "Initial AdaLoRA rank stats:",
    adalora_rank_stats(),
)

adapter_table = pd.DataFrame({
    "targeted_module":
        targeted_module_names,
})

adapter_table.to_csv(
    METRICS_DIR
    / "adalora_target_modules.csv",
    index=False,
)

display(
    adapter_table.head(30)
)


# ------------------------------------------------------------
# Feature-shape sanity test.
# ------------------------------------------------------------
model.eval()

with torch.no_grad():
    dummy = torch.zeros(
        1,
        3,
        IMAGE_SIZE,
        IMAGE_SIZE,
        device=DEVICE,
    )

    body_features = (
        model.backbone.body(dummy)
    )

    body_features[-1] = (
        model.backbone
        .deep_mamba(
            body_features[-1]
        )
    )

    ordered = OrderedDict(
        (
            str(i),
            feat,
        )
        for i, feat
        in enumerate(body_features)
    )

    raw_fpn = (
        model.backbone
        .fpn(ordered)
    )

    fused_fpn = (
        model.backbone
        .asff(raw_fpn)
    )

print(
    "Raw FPN shapes:",
    {
        key: tuple(value.shape)
        for key, value
        in raw_fpn.items()
    },
)

print(
    "ASFF output shapes:",
    {
        key: tuple(value.shape)
        for key, value
        in fused_fpn.items()
    },
)

for key in raw_fpn:
    if (
        raw_fpn[key].shape
        != fused_fpn[key].shape
    ):
        raise RuntimeError(
            f"ASFF changed FPN shape at {key}."
        )

print(
    "✅ A2 architecture + canonical AdaLoRA shape sanity check passed."
)

del dummy
del body_features
del ordered
del raw_fpn
del fused_fpn

gc.collect()
torch.cuda.empty_cache()
```

## 11. AdaLoRA optimizer, LR warm-up, cosine decay and AMP

```python
adalora_param_ids = {
    id(p)
    for p in adalora_params_list
}

trainable_ids = {
    id(p)
    for p in model.parameters()
    if p.requires_grad
}

if trainable_ids != adalora_param_ids:
    raise RuntimeError(
        "Trainable parameter set does not match AdaLoRA parameters."
    )

optimizer = torch.optim.AdamW(
    [
        {
            "params":
                adalora_params_list,

            "lr":
                LR_ADALORA,

            "name":
                "adalora",
        },
    ],

    weight_decay=
        WEIGHT_DECAY,
)


def configure_training_stage(
    epoch,
    global_step,
):
    if epoch <= LR_WARMUP_EPOCHS:
        warmup_fraction = (
            epoch
            / max(
                1,
                LR_WARMUP_EPOCHS,
            )
        )

        current_lr = (
            LR_ADALORA
            * warmup_fraction
        )

        cosine_factor = 1.0

    else:
        decay_total = max(
            1,
            MAX_EPOCHS
            - LR_WARMUP_EPOCHS,
        )

        decay_index = (
            epoch
            - LR_WARMUP_EPOCHS
            - 1
        )

        progress = (
            decay_index
            / max(
                1,
                decay_total - 1,
            )
        )

        progress = float(
            np.clip(
                progress,
                0.0,
                1.0,
            )
        )

        cosine_factor = (
            0.5
            * (
                1.0
                + np.cos(
                    np.pi
                    * progress
                )
            )
        )

        current_lr = max(
            MIN_LR,
            LR_ADALORA
            * cosine_factor,
        )

    optimizer.param_groups[0][
        "lr"
    ] = current_lr

    return {
        "stage":
            adalora_phase(
                global_step
            ),

        "cosine_factor":
            cosine_factor,

        "lr_adalora":
            current_lr,
    }


def keep_frozen_modules_in_eval():
    model.backbone.deep_mamba.eval()
    model.backbone.fpn.eval()
    model.backbone.asff.eval()
    model.head.eval()

    # PVTv2 body remains in train mode so AdaLoRA is active.
    model.backbone.body.train()


try:
    scaler = torch.amp.GradScaler(
        "cuda",
        enabled=AMP,
    )
except TypeError:
    scaler = (
        torch.cuda.amp
        .GradScaler(
            enabled=AMP,
        )
    )


def autocast_context():
    try:
        return torch.amp.autocast(
            "cuda",
            enabled=AMP,
        )
    except TypeError:
        return torch.cuda.amp.autocast(
            enabled=AMP,
        )


print("Optimizer: AdamW")

print(
    "Trainable component: canonical AdaLoRA only"
)

print(
    "AdaLoRA LR:",
    LR_ADALORA,
)

print(
    "LR warm-up epochs:",
    LR_WARMUP_EPOCHS,
)

print(
    "Initial rank:",
    ADALORA_INIT_R,
)

print(
    "Target average rank:",
    ADALORA_TARGET_R,
)

print(
    "Orthogonal regularization:",
    ADALORA_ORTH_REG_WEIGHT,
)

print(
    "Weight decay:",
    WEIGHT_DECAY,
)

print(
    "AMP:",
    AMP,
)
```

## 12. Validation helpers

```python
def move_targets_to_device(targets):
    return [
        {
            k: v.to(
                DEVICE,
                non_blocking=True,
            )
            for k, v in t.items()
        }
        for t in targets
    ]


@torch.no_grad()
def compute_validation_loss(model, loader):
    # Torchvision detection models return losses in train mode.
    # PVTv2-B0 in timm has no random augmentation here;
    # gradients remain disabled by @torch.no_grad().
    model.train()

    total_cls = 0.0
    total_box = 0.0
    total_images = 0

    start = time.perf_counter()

    for images, targets, _ in loader:
        images = [
            img.to(
                DEVICE,
                non_blocking=True,
            )
            for img in images
        ]

        targets = move_targets_to_device(targets)

        with autocast_context():
            losses = model(
                images,
                targets,
            )

        batch_n = len(images)

        total_cls += (
            float(
                losses["classification"].detach()
            )
            * batch_n
        )

        total_box += (
            float(
                losses["bbox_regression"].detach()
            )
            * batch_n
        )

        total_images += batch_n

    torch.cuda.synchronize()

    seconds = (
        time.perf_counter()
        - start
    )

    model.eval()

    cls_loss = (
        total_cls
        / max(total_images, 1)
    )

    box_loss = (
        total_box
        / max(total_images, 1)
    )

    return {
        "classification_loss": cls_loss,
        "bbox_regression_loss": box_loss,
        "total_loss": cls_loss + box_loss,
        "seconds": seconds,
    }


@torch.no_grad()
def evaluate_map(model, loader):
    model.eval()

    metric = MeanAveragePrecision(
        box_format="xyxy",
        iou_type="bbox",
        class_metrics=False,
    )

    start = time.perf_counter()

    for images, targets, _ in loader:
        images_gpu = [
            img.to(
                DEVICE,
                non_blocking=True,
            )
            for img in images
        ]

        with autocast_context():
            outputs = model(images_gpu)

        preds = [
            {
                "boxes": out["boxes"].detach().cpu(),
                "scores": out["scores"].detach().cpu(),
                "labels": out["labels"].detach().cpu(),
            }
            for out in outputs
        ]

        gts = [
            {
                "boxes": t["boxes"].detach().cpu(),
                "labels": t["labels"].detach().cpu(),
            }
            for t in targets
        ]

        metric.update(preds, gts)

    torch.cuda.synchronize()

    seconds = (
        time.perf_counter()
        - start
    )

    result = metric.compute()

    return {
        "map50": float(result["map_50"]),
        "map75": float(result["map_75"]),
        "map50_95": float(result["map"]),
        "seconds": seconds,
    }
```

### 13.1 A2-transfer sanity check and epoch-0 fallback

```python
if not RESUME:
    print(
        "Checking exact transferred A2 path on validation set..."
    )

    initial_nonzero_E = 0

    for name, parameter in (
        adalora_body.named_parameters()
    ):
        if "lora_E" in name:
            initial_nonzero_E += int(
                torch.count_nonzero(
                    parameter.detach()
                ).item()
            )

    if initial_nonzero_E != 0:
        raise RuntimeError(
            "AdaLoRA update is non-zero before training."
        )

    initial_val = evaluate_map(
        model,
        val_loader,
    )

    baseline_map = float(
        initial_val[
            "map50_95"
        ]
    )

    best_map = baseline_map
    best_epoch = 0
    epochs_without_improvement = 0
    global_step = 0

    best_state = {
        "epoch":
            0,

        "global_step":
            0,

        "baseline_map":
            baseline_map,

        "best_map50_95":
            best_map,

        "model_state": {
            k: v.detach().cpu()
            for k, v
            in model.state_dict().items()
        },

        "class_names":
            CLASS_NAMES,

        "training_config":
            training_config,

        "rank_pattern":
            current_rank_pattern(),
    }

    torch.save(
        best_state,
        BEST_PT,
    )

    initial_rank_stats = (
        adalora_rank_stats()
    )

    initial_val_df = pd.DataFrame([
        {
            "source_A2_checkpoint":
                str(A2_BEST_PT),

            "stored_A2_val_mAP50-95":
                a2_checkpoint.get(
                    "best_map50_95",
                    np.nan,
                ),

            "A3b_initial_val_mAP50":
                initial_val[
                    "map50"
                ],

            "A3b_initial_val_mAP75":
                initial_val[
                    "map75"
                ],

            "A3b_initial_val_mAP50-95":
                initial_val[
                    "map50_95"
                ],

            "baseline_required_gain":
                BASELINE_REQUIRED_GAIN,

            "adalora_target_modules":
                len(
                    targeted_module_names
                ),

            "adalora_trainable_params":
                adalora_params_count,

            "adalora_init_r":
                ADALORA_INIT_R,

            "adalora_target_r":
                ADALORA_TARGET_R,

            "initial_active_rank_total":
                initial_rank_stats[
                    "active_rank_total"
                ],
        }
    ])

    initial_val_df.to_csv(
        METRICS_DIR
        / "initial_transfer_validation.csv",
        index=False,
    )

    display(initial_val_df)

    stored_a2 = (
        a2_checkpoint.get(
            "best_map50_95"
        )
    )

    if stored_a2 is not None:
        difference = abs(
            initial_val[
                "map50_95"
            ]
            - float(stored_a2)
        )

        print(
            "Absolute difference from stored "
            f"A2 val mAP50-95: {difference:.6f}"
        )

        if difference > 0.002:
            warnings.warn(
                "Transferred A2 validation metric differs "
                "from the stored A2 metric by >0.002. "
                "Check transfer/loading before training."
            )

    print(
        "Epoch-0 A2 fallback saved with val mAP50-95:",
        round(
            best_map,
            6,
        ),
    )

    print(
        "A trained AdaLoRA checkpoint must exceed:",
        round(
            baseline_map
            + BASELINE_REQUIRED_GAIN,
            6,
        ),
        "before replacing the A2 fallback.",
    )

    random.seed(SEED)
    np.random.seed(SEED)
    torch.manual_seed(SEED)
    torch.cuda.manual_seed_all(SEED)

    print(
        "RNGs reset after transfer sanity check."
    )

else:
    print(
        "Resume mode: initial A2-transfer check skipped."
    )
```

## 14. Train — canonical AdaLoRA only

```python
from tqdm.auto import tqdm

for epoch in range(
    start_epoch,
    MAX_EPOCHS + 1,
):

    stage_info = configure_training_stage(
        epoch,
        global_step,
    )

    model.train()
    keep_frozen_modules_in_eval()

    print(
        f"\nTraining stage: {stage_info['stage']}"
    )

    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()

    epoch_start = time.perf_counter()

    train_cls_sum = 0.0
    train_box_sum = 0.0
    train_orth_sum = 0.0
    train_images = 0

    optimizer.zero_grad(
        set_to_none=True
    )

    n_steps = len(train_loader)

    train_bar = tqdm(
        enumerate(train_loader),
        total=n_steps,
        desc=f"Epoch {epoch}/{MAX_EPOCHS}",
        dynamic_ncols=True,
    )

    for step, (
        images,
        targets,
        _,
    ) in train_bar:

        images = [
            img.to(
                DEVICE,
                non_blocking=True,
            )
            for img in images
        ]

        targets = move_targets_to_device(
            targets
        )

        group_start = (
            step // ACCUM_STEPS
        ) * ACCUM_STEPS

        group_end = min(
            group_start
            + ACCUM_STEPS,
            n_steps,
        )

        group_size = (
            group_end
            - group_start
        )

        with autocast_context():
            losses = model(
                images,
                targets,
            )

            cls_loss = losses[
                "classification"
            ]

            box_loss = losses[
                "bbox_regression"
            ]

            detection_loss = (
                cls_loss
                + box_loss
            )

            orth_loss = (
                adalora_orthogonal_regularization(
                    adalora_body,
                    ADALORA_ORTH_REG_WEIGHT,
                )
            )

            optimization_loss = (
                detection_loss
                + orth_loss
            )

            if not torch.isfinite(
                optimization_loss
            ):
                raise RuntimeError(
                    f"Non-finite training loss at "
                    f"epoch {epoch}, step {step}. "
                    "Training stopped immediately."
                )

            scaled_loss = (
                optimization_loss
                / group_size
            )

        scaler.scale(
            scaled_loss
        ).backward()

        if step == group_end - 1:
            scaler.unscale_(
                optimizer
            )

            torch.nn.utils.clip_grad_norm_(
                adalora_params_list,
                GRAD_CLIP,
            )

            # Optimizer update first.
            scaler.step(
                optimizer
            )

            scaler.update()

            # AdaLoRA requires update_and_allocate() while
            # parameter gradients still exist.
            global_step += 1

            adalora_tuner.update_and_allocate(
                global_step
            )

            optimizer.zero_grad(
                set_to_none=True
            )

        batch_n = len(images)

        train_cls_sum += (
            float(
                cls_loss.detach()
            )
            * batch_n
        )

        train_box_sum += (
            float(
                box_loss.detach()
            )
            * batch_n
        )

        train_orth_sum += (
            float(
                orth_loss.detach()
            )
            * batch_n
        )

        train_images += batch_n

        train_bar.set_postfix({
            "loss":
                f"{float(detection_loss.detach()):.4f}",

            "orth":
                f"{float(orth_loss.detach()):.4f}",

            "lr":
                f"{optimizer.param_groups[0]['lr']:.2e}",

            "rank":
                f"{adalora_rank_stats()['active_rank_mean']:.1f}",

            "VRAM":
                f"{torch.cuda.memory_allocated()/1024**3:.1f}G",
        })

    torch.cuda.synchronize()

    train_seconds = (
        time.perf_counter()
        - epoch_start
    )

    train_cls = (
        train_cls_sum
        / train_images
    )

    train_box = (
        train_box_sum
        / train_images
    )

    train_orth = (
        train_orth_sum
        / train_images
    )

    train_total = (
        train_cls
        + train_box
    )

    print(
        "\nCalculating validation loss..."
    )

    val_loss_result = (
        compute_validation_loss(
            model,
            val_loader,
        )
    )

    print(
        "Calculating validation mAP..."
    )

    val_map_result = evaluate_map(
        model,
        val_loader,
    )

    total_seconds = (
        train_seconds
        + val_loss_result[
            "seconds"
        ]
        + val_map_result[
            "seconds"
        ]
    )

    peak_vram_gb = (
        torch.cuda.max_memory_allocated()
        / 1024**3
    )

    rank_stats = (
        adalora_rank_stats()
    )

    current_phase = (
        adalora_phase(
            global_step
        )
    )

    record = {
        "epoch":
            epoch,

        "training_stage":
            current_phase,

        "global_step":
            global_step,

        "train/classification_loss":
            train_cls,

        "train/bbox_regression_loss":
            train_box,

        "train/total_loss":
            train_total,

        "train/adalora_orth_loss":
            train_orth,

        "val/classification_loss":
            val_loss_result[
                "classification_loss"
            ],

        "val/bbox_regression_loss":
            val_loss_result[
                "bbox_regression_loss"
            ],

        "val/total_loss":
            val_loss_result[
                "total_loss"
            ],

        "val/mAP50":
            val_map_result[
                "map50"
            ],

        "val/mAP75":
            val_map_result[
                "map75"
            ],

        "val/mAP50-95":
            val_map_result[
                "map50_95"
            ],

        "train_seconds":
            train_seconds,

        "val_loss_seconds":
            val_loss_result[
                "seconds"
            ],

        "val_map_seconds":
            val_map_result[
                "seconds"
            ],

        "total_seconds":
            total_seconds,

        "lr":
            optimizer.param_groups[
                0
            ][
                "lr"
            ],

        "lr_adalora":
            optimizer.param_groups[
                0
            ][
                "lr"
            ],

        "adalora_active_rank_total":
            rank_stats[
                "active_rank_total"
            ],

        "adalora_active_rank_mean":
            rank_stats[
                "active_rank_mean"
            ],

        "adalora_active_rank_min":
            rank_stats[
                "active_rank_min"
            ],

        "adalora_active_rank_max":
            rank_stats[
                "active_rank_max"
            ],

        "mamba_gate":
            float(
                model.backbone
                .deep_mamba
                .gate
                .detach()
                .cpu()
            ),

        "asff_gate_P3":
            float(
                model.backbone
                .asff
                .levels[0]
                .gate
                .detach()
                .cpu()
            ),

        "asff_gate_P4":
            float(
                model.backbone
                .asff
                .levels[1]
                .gate
                .detach()
                .cpu()
            ),

        "asff_gate_P5":
            float(
                model.backbone
                .asff
                .levels[2]
                .gate
                .detach()
                .cpu()
            ),

        "peak_vram_GB":
            peak_vram_gb,
    }

    history.append(
        record
    )

    history_df = pd.DataFrame(
        history
    )

    history_df.to_csv(
        METRICS_DIR
        / "training_history.csv",
        index=False,
    )

    current_map = float(
        val_map_result[
            "map50_95"
        ]
    )

    if (
        not np.isfinite(
            current_map
        )
        or current_map < 0
    ):
        raise RuntimeError(
            f"Invalid validation mAP50-95 "
            f"at epoch {epoch}: {current_map}. "
            "Checkpoint remains available from the "
            "previous completed epoch."
        )

    if best_epoch == 0:
        improved = (
            current_map
            > baseline_map
            + BASELINE_REQUIRED_GAIN
        )
    else:
        improved = (
            current_map
            > best_map
        )

    if improved:
        best_map = current_map
        best_epoch = epoch

        epochs_without_improvement = 0

        best_state = {
            "epoch":
                epoch,

            "global_step":
                global_step,

            "baseline_map":
                baseline_map,

            "best_map50_95":
                best_map,

            "model_state": {
                k: v.detach().cpu()
                for k, v
                in model.state_dict().items()
            },

            "class_names":
                CLASS_NAMES,

            "training_config":
                training_config,

            "rank_pattern":
                current_rank_pattern(),

            "rank_stats":
                rank_stats,
        }

        torch.save(
            best_state,
            BEST_PT,
        )

    else:
        if epoch > ADALORA_INIT_EPOCHS:
            epochs_without_improvement += 1
        else:
            epochs_without_improvement = 0

    last_state = {
        "epoch":
            epoch,

        "global_step":
            global_step,

        "baseline_map":
            baseline_map,

        "model_state":
            model.state_dict(),

        "optimizer_state":
            optimizer.state_dict(),

        "scaler_state":
            scaler.state_dict(),

        "best_map":
            best_map,

        "best_epoch":
            best_epoch,

        "epochs_without_improvement":
            epochs_without_improvement,

        "history":
            history,

        "training_config":
            training_config,

        "rank_pattern":
            current_rank_pattern(),

        "rankallocator_state":
            capture_rankallocator_state(),

        "rank_stats":
            rank_stats,
    }

    torch.save(
        last_state,
        LAST_PT,
    )

    print(
        "\n"
        + "=" * 72
    )

    print(
        f"EPOCH {epoch}/{MAX_EPOCHS} COMPLETE"
    )

    print(
        f"Stage         : "
        f"{current_phase}"
    )

    print("-" * 72)

    print(
        f"Train Loss    : "
        f"{train_total:.4f} "
        f"(Cls {train_cls:.4f} | "
        f"Box {train_box:.4f})"
    )

    print(
        f"Orth Reg      : "
        f"{train_orth:.4f}"
    )

    print(
        f"Val Loss      : "
        f"{val_loss_result['total_loss']:.4f}"
    )

    print(
        f"Val mAP@50    : "
        f"{val_map_result['map50']:.4f}"
    )

    print(
        f"Val mAP@75    : "
        f"{val_map_result['map75']:.4f}"
    )

    print(
        f"Val mAP@50-95 : "
        f"{current_map:.4f}"
    )

    print("-" * 72)

    print(
        f"AdaLoRA LR    : "
        f"{optimizer.param_groups[0]['lr']:.2e}"
    )

    print(
        "Active ranks  : "
        f"mean={rank_stats['active_rank_mean']:.2f} | "
        f"min={rank_stats['active_rank_min']} | "
        f"max={rank_stats['active_rank_max']} | "
        f"total={rank_stats['active_rank_total']}"
    )

    print(
        f"Global step   : "
        f"{global_step}/{TOTAL_UPDATE_STEPS}"
    )

    print(
        f"Train time    : "
        f"{train_seconds/60:.2f} min"
    )

    print(
        f"Val-loss time : "
        f"{val_loss_result['seconds']/60:.2f} min"
    )

    print(
        f"Val-mAP time  : "
        f"{val_map_result['seconds']/60:.2f} min"
    )

    print(
        f"Total epoch   : "
        f"{total_seconds/60:.2f} min"
    )

    print(
        f"Peak VRAM     : "
        f"{peak_vram_gb:.2f} GB"
    )

    print("-" * 72)

    print(
        f"A2 baseline   : "
        f"{baseline_map:.4f}"
    )

    print(
        f"Best epoch    : "
        f"{best_epoch}"
    )

    print(
        f"Best mAP50-95 : "
        f"{best_map:.4f}"
    )

    print(
        f"Early stopping: "
        f"{epochs_without_improvement}/{PATIENCE}"
    )

    print("=" * 72)

    if (
        epoch > ADALORA_INIT_EPOCHS
        and epochs_without_improvement
        >= PATIENCE
    ):
        print(
            f"\nEarly stopping triggered "
            f"at epoch {epoch}."
        )

        break

print("\nTRAINING FINISHED")

print(
    "Best epoch:",
    best_epoch,
)

print(
    "A2 baseline validation mAP@50-95:",
    round(
        baseline_map,
        4,
    ),
)

print(
    "Best validation mAP@50-95:",
    round(
        best_map,
        4,
    ),
)

if best_epoch == 0:
    print(
        "AdaLoRA did not clear the validation-improvement margin. "
        "The final best checkpoint remains the exact A2 fallback."
    )
else:
    print(
        "AdaLoRA cleared the validation-improvement margin."
    )
```

## 19. Collect test predictions once

```python
@torch.no_grad()
def collect_predictions(
    model,
    loader,
):
    model.eval()

    predictions = []
    targets_all = []
    paths_all = []

    total_forward_seconds = 0.0
    total_images = 0

    for images, targets, paths in loader:
        images_gpu = [
            img.to(
                DEVICE,
                non_blocking=True,
            )
            for img in images
        ]

        torch.cuda.synchronize()
        start = time.perf_counter()

        with autocast_context():
            outputs = model(
                images_gpu
            )

        torch.cuda.synchronize()

        total_forward_seconds += (
            time.perf_counter()
            - start
        )

        total_images += len(images)

        for out in outputs:
            predictions.append({
                "boxes":
                    out["boxes"].detach().cpu(),

                "scores":
                    out["scores"].detach().cpu(),

                "labels":
                    out["labels"].detach().cpu(),
            })

        for target in targets:
            targets_all.append({
                "boxes":
                    target["boxes"].detach().cpu(),

                "labels":
                    target["labels"].detach().cpu(),
            })

        paths_all.extend(paths)

    inference_ms = (
        total_forward_seconds
        / max(total_images, 1)
        * 1000
    )

    return (
        predictions,
        targets_all,
        paths_all,
        inference_ms,
    )


(
    test_predictions,
    test_targets,
    test_paths,
    inference_ms,
) = collect_predictions(
    model,
    test_loader,
)

fps = (
    1000 / inference_ms
    if inference_ms > 0
    else np.nan
)

print(
    "Test images:",
    len(test_predictions),
)

print(
    f"Model forward time: "
    f"{inference_ms:.3f} ms/image"
)

print(
    f"FPS: {fps:.2f}"
)
```

## 21. Precision, Recall, F1, matched box IoU and Dice

```python
def class_aware_matches(
    pred,
    gt,
    conf_threshold=0.25,
    iou_threshold=0.50,
):
    keep = (
        pred["scores"]
        >= conf_threshold
    )

    p_boxes = pred["boxes"][keep]
    p_labels = pred["labels"][keep]

    g_boxes = gt["boxes"]
    g_labels = gt["labels"]

    if (
        len(p_boxes) == 0
        or len(g_boxes) == 0
    ):
        return (
            [],
            p_boxes,
            p_labels,
        )

    ious = box_iou(
        p_boxes,
        g_boxes,
    )

    candidates = []

    for pi in range(
        len(p_boxes)
    ):
        for gi in range(
            len(g_boxes)
        ):
            if (
                int(p_labels[pi])
                != int(g_labels[gi])
            ):
                continue

            iou = float(
                ious[pi, gi]
            )

            if (
                iou
                >= iou_threshold
            ):
                candidates.append(
                    (
                        iou,
                        pi,
                        gi,
                    )
                )

    candidates.sort(
        reverse=True
    )

    used_pred = set()
    used_gt = set()
    matches = []

    for iou, pi, gi in candidates:
        if (
            pi in used_pred
            or gi in used_gt
        ):
            continue

        used_pred.add(pi)
        used_gt.add(gi)

        matches.append(
            (
                pi,
                gi,
                iou,
            )
        )

    return (
        matches,
        p_boxes,
        p_labels,
    )


overall_tp = 0
overall_fp = 0
overall_fn = 0

matched_ious = []

per_class_counts = defaultdict(
    lambda: {
        "tp": 0,
        "fp": 0,
        "fn": 0,
        "ious": [],
    }
)

for pred, gt in zip(
    test_predictions,
    test_targets,
):
    (
        matches,
        p_boxes,
        p_labels,
    ) = class_aware_matches(
        pred,
        gt,
        CONF_THRESHOLD,
        MATCH_IOU_THRESHOLD,
    )

    matched_p = {
        m[0]
        for m in matches
    }

    matched_g = {
        m[1]
        for m in matches
    }

    tp = len(matches)
    fp = len(p_boxes) - tp
    fn = len(gt["boxes"]) - tp

    overall_tp += tp
    overall_fp += fp
    overall_fn += fn

    for (
        pi,
        gi,
        iou,
    ) in matches:
        cls_id = int(
            gt["labels"][gi]
        )

        matched_ious.append(
            iou
        )

        per_class_counts[
            cls_id
        ]["tp"] += 1

        per_class_counts[
            cls_id
        ]["ious"].append(
            iou
        )

    for pi, cls_id in enumerate(
        p_labels.tolist()
    ):
        if pi not in matched_p:
            per_class_counts[
                int(cls_id)
            ]["fp"] += 1

    for gi, cls_id in enumerate(
        gt["labels"].tolist()
    ):
        if gi not in matched_g:
            per_class_counts[
                int(cls_id)
            ]["fn"] += 1


test_precision = (
    overall_tp
    / (overall_tp + overall_fp)
    if (
        overall_tp + overall_fp
    )
    else 0.0
)

test_recall = (
    overall_tp
    / (overall_tp + overall_fn)
    if (
        overall_tp + overall_fn
    )
    else 0.0
)

test_f1 = (
    2
    * test_precision
    * test_recall
    / (
        test_precision
        + test_recall
    )
    if (
        test_precision
        + test_recall
    )
    else 0.0
)

mean_iou = (
    float(
        np.mean(matched_ious)
    )
    if matched_ious
    else np.nan
)

matched_dices = [
    2 * iou / (1 + iou)
    for iou in matched_ious
]

mean_dice = (
    float(
        np.mean(matched_dices)
    )
    if matched_dices
    else np.nan
)

print(
    f"Precision : "
    f"{test_precision:.4f}"
)

print(
    f"Recall    : "
    f"{test_recall:.4f}"
)

print(
    f"F1        : "
    f"{test_f1:.4f}"
)

print(
    f"TP / FP / FN: "
    f"{overall_tp} / "
    f"{overall_fp} / "
    f"{overall_fn}"
)

print(
    f"Mean matched box IoU : "
    f"{mean_iou:.4f}"
)

print(
    f"Mean matched box Dice: "
    f"{mean_dice:.4f}"
)
```

## 22. Per-class test performance

```python
map95_by_class = {}
map50_by_class = {}

classes95 = (
    map_result["classes"]
    .cpu()
    .numpy()
    .astype(int)
)

values95 = (
    map_result["map_per_class"]
    .cpu()
    .numpy()
)

for cls_id, value in zip(
    classes95,
    values95,
):
    map95_by_class[
        int(cls_id)
    ] = (
        float(value)
        if value >= 0
        else np.nan
    )

classes50 = (
    map50_result["classes"]
    .cpu()
    .numpy()
    .astype(int)
)

values50 = (
    map50_result["map_per_class"]
    .cpu()
    .numpy()
)

for cls_id, value in zip(
    classes50,
    values50,
):
    map50_by_class[
        int(cls_id)
    ] = (
        float(value)
        if value >= 0
        else np.nan
    )


class_image_counts = defaultdict(
    set
)

class_instance_counts = defaultdict(
    int
)

for image_idx, target in enumerate(
    test_targets
):
    for cls_id in target[
        "labels"
    ].tolist():
        class_image_counts[
            int(cls_id)
        ].add(image_idx)

        class_instance_counts[
            int(cls_id)
        ] += 1


rows = []

for cls_id, class_name in enumerate(
    CLASS_NAMES
):
    stats = per_class_counts[
        cls_id
    ]

    tp = stats["tp"]
    fp = stats["fp"]
    fn = stats["fn"]

    p = (
        tp / (tp + fp)
        if tp + fp
        else np.nan
    )

    r = (
        tp / (tp + fn)
        if tp + fn
        else np.nan
    )

    f1 = (
        2 * p * r / (p + r)
        if (
            np.isfinite(p)
            and np.isfinite(r)
            and p + r > 0
        )
        else np.nan
    )

    cls_iou = (
        float(
            np.mean(
                stats["ious"]
            )
        )
        if stats["ious"]
        else np.nan
    )

    cls_dice = (
        2
        * cls_iou
        / (1 + cls_iou)
        if np.isfinite(
            cls_iou
        )
        else np.nan
    )

    rows.append({
        "Class": class_name,
        "Images":
            len(
                class_image_counts[
                    cls_id
                ]
            ),
        "Instances":
            class_instance_counts[
                cls_id
            ],
        "Box-P": p,
        "Box-R": r,
        "Box-F1": f1,
        "mAP50":
            map50_by_class.get(
                cls_id,
                np.nan,
            ),
        "mAP50-95":
            map95_by_class.get(
                cls_id,
                np.nan,
            ),
        "Mean Box IoU":
            cls_iou,
        "Mean Box Dice":
            cls_dice,
    })


per_class_df = pd.DataFrame(
    rows
).round(4)

per_class_df.to_csv(
    METRICS_DIR
    / "final_per_class_summary.csv",
    index=False,
)

display(per_class_df)
```

## 23. Raw and normalized confusion matrices

```python
N = NUM_CLASSES
BG = N

confusion = np.zeros(
    (N + 1, N + 1),
    dtype=np.int64,
)

for pred, gt in zip(
    test_predictions,
    test_targets,
):
    keep = (
        pred["scores"]
        >= CONF_THRESHOLD
    )

    p_boxes = pred["boxes"][keep]
    p_labels = pred["labels"][keep]

    g_boxes = gt["boxes"]
    g_labels = gt["labels"]

    used_p = set()
    used_g = set()

    if (
        len(p_boxes)
        and len(g_boxes)
    ):
        ious = box_iou(
            p_boxes,
            g_boxes,
        )

        candidates = []

        for pi in range(
            len(p_boxes)
        ):
            for gi in range(
                len(g_boxes)
            ):
                iou = float(
                    ious[pi, gi]
                )

                if (
                    iou
                    >= MATCH_IOU_THRESHOLD
                ):
                    candidates.append(
                        (
                            iou,
                            pi,
                            gi,
                        )
                    )

        candidates.sort(
            reverse=True
        )

        for (
            iou,
            pi,
            gi,
        ) in candidates:
            if (
                pi in used_p
                or gi in used_g
            ):
                continue

            used_p.add(pi)
            used_g.add(gi)

            confusion[
                int(g_labels[gi]),
                int(p_labels[pi]),
            ] += 1

    for gi, cls_id in enumerate(
        g_labels.tolist()
    ):
        if gi not in used_g:
            confusion[
                int(cls_id),
                BG,
            ] += 1

    for pi, cls_id in enumerate(
        p_labels.tolist()
    ):
        if pi not in used_p:
            confusion[
                BG,
                int(cls_id),
            ] += 1


labels_cm = (
    CLASS_NAMES
    + ["background"]
)

row_sums = confusion.sum(
    axis=1,
    keepdims=True,
)

normalized = np.divide(
    confusion,
    row_sums,
    out=np.zeros_like(
        confusion,
        dtype=float,
    ),
    where=row_sums != 0,
)

pd.DataFrame(
    confusion,
    index=labels_cm,
    columns=labels_cm,
).to_csv(
    METRICS_DIR
    / "confusion_matrix.csv"
)

pd.DataFrame(
    normalized,
    index=labels_cm,
    columns=labels_cm,
).to_csv(
    METRICS_DIR
    / "confusion_matrix_normalized.csv"
)


def plot_confusion(
    matrix,
    title,
    filename,
    fmt,
):
    fig, ax = plt.subplots(
        figsize=(11, 9)
    )

    ax.imshow(matrix)

    ax.set_xticks(
        range(len(labels_cm))
    )

    ax.set_yticks(
        range(len(labels_cm))
    )

    ax.set_xticklabels(
        labels_cm,
        rotation=45,
        ha="right",
    )

    ax.set_yticklabels(
        labels_cm
    )

    ax.set_xlabel(
        "Predicted class"
    )

    ax.set_ylabel(
        "Ground-truth class"
    )

    ax.set_title(title)

    for i in range(
        matrix.shape[0]
    ):
        for j in range(
            matrix.shape[1]
        ):
            value = matrix[i, j]

            text = (
                f"{value:.2f}"
                if fmt == "float"
                else str(int(value))
            )

            if value != 0:
                ax.text(
                    j,
                    i,
                    text,
                    ha="center",
                    va="center",
                    fontsize=7,
                )

    plt.tight_layout()

    plt.savefig(
        PLOTS_DIR / filename,
        dpi=300,
        bbox_inches="tight",
    )

    plt.show()


plot_confusion(
    confusion,
    "PVTv2-B0 + RetinaNet Confusion Matrix",
    "confusion_matrix.png",
    "int",
)

plot_confusion(
    normalized,
    "PVTv2-B0 + RetinaNet Normalized Confusion Matrix",
    "confusion_matrix_normalized.png",
    "float",
)
```
